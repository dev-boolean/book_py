### Marco teórico.

#### 1. El mercado laboral: oferta, demanda y mecanismos de ajuste

El mercado laboral se define como el espacio económico en el que interactúan la oferta y la demanda de trabajo. El salario actúa como el mecanismo de ajuste que equilibra estas fuerzas: cuando la demanda de trabajo supera la oferta, los salarios tienden a subir, incentivando una mayor participación laboral; cuando ocurre lo contrario, el desempleo surge como consecuencia de un exceso de oferta (Mankiw, 2015). Sin embargo, este ajuste rara vez es inmediato ni perfecto, dado que existen fricciones e instituciones que afectan o distorsionan el equilibrio.

Desde la perspectiva keynesiana, el desempleo puede persistir incluso en equilibrio cuando la demanda agregada es insuficiente para absorber toda la oferta laboral disponible (Keynes, 1936). Esta visión contrasta con el enfoque neoclásico, que postula que el mercado tiende al pleno empleo en el largo plazo, y con el concepto de tasa natural de desempleo desarrollado por Friedman (1968), según el cual existe un nivel de desempleo estructural y friccional que la política monetaria no puede reducir de forma permanente.

#### 2. Fricciones, búsqueda y emparejamiento

Los modelos de búsqueda y emparejamiento (*search and matching*), desarrollados principalmente por Pissarides (2000) y Diamond, ofrecen un marco analítico para entender por qué el desempleo persiste incluso en economías dinámicas. En estos modelos, trabajadores y empresas incurren en costos de búsqueda y el proceso de emparejamiento no es instantáneo, lo que genera desempleo friccional. La curva de Beveridge —que relaciona inversamente las vacantes y el desempleo— constituye una representación empírica central de este enfoque (Blanchard & Diamond, 1989).

Adicionalmente, el fenómeno de histéresis, documentado por Blanchard y Summers (1987), sugiere que choques temporales sobre el desempleo pueden generar efectos permanentes sobre la tasa natural, especialmente cuando los trabajadores desempleados pierden capital humano o son excluidos del proceso de negociación salarial.

#### 3. Indicadores del mercado laboral colombiano

Para un mejor entendimiento del comportamiento del mercado laboral colombiano, el DANE utiliza definiciones establecidas por la Organización Internacional del Trabajo (OIT). Bajo estos conceptos, la población en edad de trabajar (PET) se divide en población económicamente activa (PEA o fuerza de trabajo) y población económicamente inactiva. A su vez, la PEA se subdivide en ocupados y desocupados, lo que da lugar a los tres indicadores centrales de este estudio:

- La **tasa global de participación** mide la proporción de la PET que hace parte de la fuerza laboral.
- La **tasa de ocupación** indica el porcentaje de la PET que se encuentra efectivamente trabajando.
- La **tasa de desempleo** señala la proporción de la fuerza de trabajo que está disponible y buscando empleo pero no lo encuentra.

Estos tres indicadores no son independientes entre sí y deben analizarse de forma conjunta para comprender la dinámica del mercado laboral.

#### 4. Series de tiempo y modelado econométrico

Desde el enfoque de las series de tiempo, el mercado laboral presenta componentes que lo hacen complejo: tendencias a largo plazo asociadas a cambios demográficos, ciclos y choques económicos, y estacionalidades derivadas de contrataciones en ciertas épocas del año. Para su análisis riguroso se emplean modelos ARIMA y, más recientemente, modelos de machine learning aplicados a la predicción de series temporales económicas (Hyndman & Athanasopoulos, 2021).

En el contexto colombiano, trabajos como el de Arango, Herrera y Posada (2008) han documentado la persistencia del desempleo y la importancia de factores institucionales en su determinación, mientras que el Banco de la República ha señalado reiteradamente el impacto diferencial de los choques externos sobre el empleo. El periodo 2020–2021 generó una ruptura estructural que cualquier modelo predictivo debe contemplar con especial atención.

- Una vez analizada la distribución de las variables mediante histogramas y detectados posibles valores atípicos con boxplot, se emplea el gráfico de dispersión con el fin de visualizar cada observación individualmente.
````python
Q1      = MLC['tasa_desempleo_nacional'].quantile(0.25)
Q3      = MLC['tasa_desempleo_nacional'].quantile(0.75)
IQR_val = Q3 - Q1
lim_inf = Q1 - 1.5 * IQR_val
lim_sup = Q3 + 1.5 * IQR_val

outliers = (MLC['tasa_desempleo_nacional'] < lim_inf) | (MLC['tasa_desempleo_nacional'] > lim_sup)

fig, ax = plt.subplots(figsize=(10, 5))

ax.scatter(MLC['fecha'], MLC['tasa_desempleo_nacional'],
           color='black', s=16)
ax.scatter(MLC.loc[outliers, 'fecha'],
           MLC.loc[outliers, 'tasa_desempleo_nacional'],
           color='red', s=30, zorder=5)

ax.set_title('Dispersión de la tasa de desempleo nacional')
ax.set_xlabel('Fecha')
ax.set_ylabel('Tasa de desempleo (%)')
plt.tight_layout()
plt.show()
````

### Analisis bivariado.

Después de analizar cada variable de forma individual y de explorar algunas relaciones mediante gráficos de dispersion, resulta util complementar con una matriz de correlación que ayudará a evaluar de manera conjunta la fuerza y la dirección de la relación lineal entre los indicadores del mercado laboral. - Se hace la prueba ADF (Augmented Dickey–Fuller). Esta herramienta nos ayuda a detectar estacionariedad. La razon primordial es determinar si una serie es estacionaria o si tiene una raíz unitaria.
Esto importa porque la mayoría de los modelos estadísticos (como ARIMA) asumen que las propiedades estadísticas de la serie son constantes en el tiempo. Si la serie no es estacionaria, los resultados pueden ser sesgados.

Si $p \le \alpha$ (0.05): Rechaza la hipótesis nula. La serie es estacionaria.

Si $p > \alpha$: No se rechaza la hipótesis nula. La serie es no estacionaria (tiene raíz unitaria).

Veamos a continuación con las variables tratadas.

````python
from statsmodels.tsa.stattools import adfuller

cols_adf = {
    'tasa_global_participacion_area':     'participacion_area',
    'tasa_global_participacion_nacional': 'participacion_nacional',
    'tasa_desempleo_area':                'desempleo_area',
    'tasa_desempleo_nacional':            'desempleo_nacional',
    'tasa_ocupacion_area':                'ocupacion_area',
    'tasa_ocupacion_nacional':            'ocupacion_nacional',
}

resultados_adf = []
for col, nombre in cols_adf.items():
    p_value = adfuller(MLC[col].dropna(), autolag='AIC')[1]
    resultados_adf.append({
        'Variable':     nombre,
        'p_value':      round(p_value, 4),
        'Estacionaria': 'Sí' if p_value < 0.05 else 'No',
    })

pd.DataFrame(resultados_adf)
````

A partir de los resultados obtenidos en la prueba Augmented Dickey-Fuller (ADF), se observa que todas las variables presentan p-values superiores a 0.05.
Esto implica que no se rechaza la hipótesis nula de presencia de raíz unitaria, por lo que ninguna de las series es estacionaria en niveles.

# Modelado Predictivo: SARIMA

## Justificación del modelo seleccionado

El análisis exploratorio desarrollado en la sección anterior dejó evidencia clara sobre las características estructurales de la serie `tasa_desempleo_nacional`: no estacionariedad en niveles (p-valor ADF = 0.069 > 0.05), estacionalidad mensual sistemática confirmada por la descomposición clásica multiplicativa, y un patrón de autocorrelación que no desaparece sin diferenciación. Estas condiciones descartan el uso de un ARIMA simple.

Un modelo **ARIMA(p,d,q)** clásico solo contempla diferenciación regular pero no captura la dependencia estacional periódica (cada 12 meses). Al ajustar ARIMA(1,1,1) sobre la serie de entrenamiento se obtiene un **AIC ≈ 895 y RMSE = 1.53**, resultados claramente inferiores a cualquier especificación estacional.

El modelo apropiado para esta serie es el **SARIMA(p,d,q)(P,D,Q)[s]**, también llamado modelo ARIMA estacional o modelo de Box-Jenkins estacional, cuya forma general es:

$$\Phi_P(B^{12})\,\phi_p(B)\,(1-B)^d(1-B^{12})^D\,y_t = \Theta_Q(B^{12})\,\theta_q(B)\,\varepsilon_t$$

donde $B$ es el operador de rezago, $d$ es el orden de diferenciación regular, $D$ es el orden de diferenciación estacional y $s = 12$ para datos mensuales.

## Comparación de especificaciones

Se evaluaron cinco modelos candidatos usando los primeros 270 meses como entrenamiento y los 30 restantes como validación:

````python
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.statespace.sarimax import SARIMAX
import numpy as np
import pandas as pd

# Serie ya definida como ts_desempleo_nacional en celdas anteriores
train = ts_desempleo_nacional.iloc[:270]
test  = ts_desempleo_nacional.iloc[270:]

candidatos = {
    'ARIMA(1,1,1)':             {'order': (1,1,1), 'seasonal_order': None},
    'SARIMA(0,1,1)(0,1,1)[12]': {'order': (0,1,1), 'seasonal_order': (0,1,1,12)},
    'SARIMA(1,1,1)(0,1,1)[12]': {'order': (1,1,1), 'seasonal_order': (0,1,1,12)},
    'SARIMA(1,1,1)(1,1,1)[12]': {'order': (1,1,1), 'seasonal_order': (1,1,1,12)},
    'SARIMA(2,1,2)(1,1,1)[12]': {'order': (2,1,2), 'seasonal_order': (1,1,1,12)},
}

resultados_comp = []
for nombre, cfg in candidatos.items():
    if cfg['seasonal_order'] is None:
        m = ARIMA(train, order=cfg['order']).fit()
    else:
        m = SARIMAX(train, order=cfg['order'],
                    seasonal_order=cfg['seasonal_order']).fit(disp=False)
    fc   = m.forecast(steps=len(test))
    rmse = np.sqrt(np.mean((fc.values - test.values)**2))
    resultados_comp.append({
        'Modelo': nombre,
        'AIC':    round(m.aic, 2),
        'BIC':    round(m.bic, 2),
        'RMSE':   round(rmse, 4),
    })

tabla_comp = pd.DataFrame(resultados_comp)
tabla_comp
````

En este split puntual, el **SARIMA(2,1,2)(1,1,1)[12]** obtiene el menor RMSE de validación (0.835) frente a especificaciones más simples. Sin embargo, un resultado basado en un único corte train/test de 30 observaciones puede ser engañoso si la complejidad adicional no está realmente respaldada por los datos. Antes de fijar el modelo final se verifican dos cosas:

1. Si los parámetros de cada especificación son estadísticamente significativos.
2. Si el ranking de RMSE se mantiene al validar con **múltiples orígenes en el tiempo** (walk-forward), no solo con el corte puntual 270/30.

````python
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.statespace.sarimax import SARIMAX

def ajustar(y, cfg):
    if cfg['seasonal_order'] is None:
        return ARIMA(y, order=cfg['order']).fit()
    return SARIMAX(y, order=cfg['order'], seasonal_order=cfg['seasonal_order']).fit(disp=False)

filas_sig = []
for nombre, cfg in candidatos.items():
    m  = ajustar(train, cfg)
    pv = m.pvalues
    filas_sig.append({
        'Modelo':                       nombre,
        'n_parametros':                 len(pv),
        'n_significativos (p<0.05)':    int((pv < 0.05).sum()),
        'parametros_no_significativos': ', '.join(pv.index[pv >= 0.05]) or '-',
    })

tabla_sig = pd.DataFrame(filas_sig)
tabla_sig
````

La tabla anterior muestra que el modelo con menor RMSE, **SARIMA(2,1,2)(1,1,1)[12]**, no tiene ningún coeficiente estadísticamente significativo (0 de 7 parámetros, todos con p > 0.19) al ajustarse sobre el mismo tramo de entrenamiento usado para compararlo. En contraste, **SARIMA(0,1,1)(0,1,1)[12]** tiene sus 3 parámetros significativos (p < 0.001). Esto es una señal clara de sobreparametrización: el modelo más complejo no está identificado de forma robusta por los datos, aunque coincida en ajustar mejor ese split puntual.

Para confirmar si la ventaja de RMSE es real o un artefacto del corte, se repite la comparación con **validación walk-forward**: se reentrena cada modelo en varios orígenes distintos y se pronostican los siguientes 12 meses en cada caso, promediando el error resultante.

````python
h = 12
origenes = [180, 204, 216, 252, 264, 276]

resultados_wf = {nombre: [] for nombre in candidatos}
for o in origenes:
    y_train_o = ts_desempleo_nacional.iloc[:o]
    y_test_o  = ts_desempleo_nacional.iloc[o:o+h]
    for nombre, cfg in candidatos.items():
        m      = ajustar(y_train_o, cfg)
        fc     = m.forecast(steps=h)
        rmse_o = np.sqrt(np.mean((fc.values - y_test_o.values)**2))
        resultados_wf[nombre].append(rmse_o)

tabla_wf = pd.DataFrame(
    resultados_wf,
    index=[ts_desempleo_nacional.index[o].strftime('%Y-%m') for o in origenes]
)
tabla_wf.loc['Promedio'] = tabla_wf.mean()
tabla_wf.loc['Mediana']  = tabla_wf.median()
tabla_wf.round(3)
````

Al validar con 6 orígenes distintos en lugar de un único corte, el ranking se invierte: **SARIMA(2,1,2)(1,1,1)[12]** pasa a ser el **peor** de los cuatro candidatos estacionales tanto en RMSE promedio como en mediana, mientras que especificaciones más simples como SARIMA(1,1,1)(1,1,1)[12] y **SARIMA(0,1,1)(0,1,1)[12]** resultan consistentemente más precisas y estables a través del tiempo.

Combinando los tres criterios —RMSE walk-forward, AIC/BIC y significancia de los coeficientes—, el mejor balance lo ofrece **SARIMA(0,1,1)(0,1,1)[12]**: su precisión es prácticamente idéntica a la del mejor candidato en el backtest (diferencia de ~3% en RMSE promedio), tiene el mejor AIC y BIC de toda la comparación, y es el único donde el 100% de los parámetros son significativos. Por parsimonia y robustez estadística, **este es el modelo que se adopta como definitivo**, en lugar del SARIMA(2,1,2)(1,1,1)[12] que sugería, de forma no robusta, el RMSE de un único split.

## Diagnóstico de residuos

````python
from statsmodels.stats.diagnostic import acorr_ljungbox

lb_results = acorr_ljungbox(residuos.dropna(), lags=[12, 24], return_df=True)

tabla_lb = pd.DataFrame({
    'Rezagos':      [12, 24],
    'Estadistico':  lb_results['lb_stat'].round(3).values,
    'p_valor':      lb_results['lb_pvalue'].round(4).values,
    'Conclusion':   ['No se rechaza H0 (ruido blanco)'
                     if p > 0.05 else 'Se rechaza H0'
                     for p in lb_results['lb_pvalue'].values]
})

tabla_lb
````

## Una identidad contable que hay que evitar

Las tres tasas del mercado laboral no son independientes entre sí. La tasa de desempleo es la proporción de la fuerza de trabajo que no tiene empleo, de modo que se puede escribir como

$$TD = \frac{PEA - OCU}{PEA} = 1 - \frac{OCU}{PEA} = 1 - \frac{TO}{TGP}$$

donde TO es la tasa de ocupación y TGP la tasa global de participación. Si esto se cumple en los datos, usar la ocupación y la participación nacionales del mismo mes para explicar el desempleo nacional no sería predecir nada: el modelo solo tendría que despejar una ecuación. Lo comprobamos con los datos:

````python
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import TimeSeriesSplit

from src import config, targets, plots
from src import analysis as A
from src import optim_analysis as OA
from src import stats_tests as ST
from src import interpretability as I
from src.data import assemble_dataset, load_raw
from src.models import CLASSIFIERS, REGRESSORS, CLASSIFIER_BY_NAME, REGRESSOR_BY_NAME
from src.nested_cv import run_nested_cv

# Se calcula 1 - TO/TGP con las variables nacionales y se compara con la tasa de desempleo nacional
datos = load_raw()
implicita = 1 - datos['tasa_ocupacion_nacional'] / datos['tasa_global_participacion_nacional']
correlacion = np.corrcoef(implicita, datos['tasa_desempleo_nacional'] / 100)[0, 1]
print(f'Correlación entre 1 - TO/TGP y la tasa de desempleo nacional: {correlacion:.7f}')
````

La correlación es de 0.9999993, es decir, la identidad se cumple casi a la perfección. Por esa razón la ocupación y la participación nacionales no se usan como variables explicativas. En cambio, los indicadores de las 13 áreas metropolitanas sí se conservan: corresponden a una población distinta (las ciudades principales) y no quedan amarrados a esa identidad. De hecho, en el análisis bivariado ya se había visto que el desempleo de área y el nacional se mueven casi juntos (correlación de 0.96), lo que los hace buenos candidatos como predictores y permite poner a prueba la primera hipótesis secundaria.

## Una validación que respeta el orden del tiempo

La validación cruzada tampoco puede hacerse mezclando los meses al azar. Como las variables incluyen rezagos y promedios móviles, el mes *t* y el mes *t+1* comparten casi toda su información (el segundo rezago de uno es el primer rezago del otro), y un K-fold aleatorio dejaría meses casi idénticos repartidos entre entrenamiento y prueba (Bergmeir & Benítez, 2012). El resultado sería una métrica demasiado optimista, que es justo el sesgo de selección que se busca evitar con la validación anidada (Cawley & Talbot, 2010).

Por eso se usó una validación anidada con ventana expansiva. El bucle externo, de 5 pliegues, mide qué tan bien generaliza el modelo ya optimizado; dentro de cada tramo de entrenamiento, un bucle interno de 3 pliegues es el que usa el optimizador para elegir los hiperparámetros. En los dos bucles la prueba queda siempre después del entrenamiento, con un mes de separación entre ambos para que los rezagos no se crucen en la frontera.

````python
particiones = list(TimeSeriesSplit(n_splits=config.N_SPLITS_OUTER, gap=config.CV_GAP).split(ds.X))
fig = plots.plot_walk_forward(fechas_objetivo, particiones)
plt.show()

# Tamaño y composición de cada pliegue externo
etiquetas = ds.y_clf_labels
filas = []
for i, (entrenamiento, prueba) in enumerate(particiones):
    filas.append({
        'Pliegue': i + 1,
        'Meses de entrenamiento': len(entrenamiento),
        'Prueba desde': fechas_objetivo.iloc[prueba[0]].strftime('%Y-%m'),
        'Prueba hasta': fechas_objetivo.iloc[prueba[-1]].strftime('%Y-%m'),
        'Bajo (entrenamiento)': int((etiquetas.iloc[entrenamiento] == 'Bajo').sum()),
        'Alto (entrenamiento)': int((etiquetas.iloc[entrenamiento] == 'Alto').sum()),
        'Alto (prueba)': int((etiquetas.iloc[prueba] == 'Alto').sum()),
    })
pd.DataFrame(filas).set_index('Pliegue')
````

Con 288 observaciones, 5 pliegues externos dejan pruebas de 48 meses (cuatro años) y un primer entrenamiento de 47 meses; más pliegues habrían dado pruebas demasiado cortas y menos habrían entregado muy pocas estimaciones del desempeño. En el bucle interno se usaron 3 pliegues para que los tramos de validación siguieran siendo razonables incluso en el primer pliegue externo. La métrica que se reporta siempre es la del bucle externo, nunca el mejor valor que encontró la búsqueda en el bucle interno.

La tabla muestra algo que condiciona los resultados: las clases no están repartidas parejo en el tiempo. ## Modelos y espacios de búsqueda

Los siete modelos de cada tarea son los pedidos en la actividad. En clasificación: KNN, Naive Bayes (gaussiano), regresión logística (con regularización L1 y L2), árbol de decisión, Random Forest, XGBoost y SVM. En regresión: KNN, Ridge, Lasso, árbol de decisión, Random Forest, XGBoost y SVR. Se usaron las implementaciones de scikit-learn (Pedregosa et al., 2011) y de XGBoost (Chen & Guestrin, 2016). A todos se les define el mismo espacio de búsqueda para los cuatro optimizadores, de manera que ninguno tenga ventaja por buscar en una región distinta.

Los rangos no son arbitrarios. Los parámetros de regularización (`C`, `alpha`, `gamma`, `var_smoothing` y la tasa de aprendizaje) se exploran en escala logarítmica, porque sus efectos cambian por órdenes de magnitud. Los parámetros de complejidad se acotan según el tamaño de los datos: con 288 observaciones no tiene sentido un árbol de más de 20 niveles ni un KNN con más de 30 vecinos. La siguiente tabla resume el espacio de cada modelo. Una limitación del diseño es que, con 17 variables, `sqrt` y `log2` dan ambas 4 variables por división, así que en los bosques `max_features` en la práctica no cambia el modelo; se dejó igual para no alterar el espacio definido desde el inicio.

````python
filas = []
for tarea, modelos in [('Clasificación', CLASSIFIERS), ('Regresión', REGRESSORS)]:
    for modelo in modelos:
        for h in modelo.space:
            if h.kind == 'cat':
                rango, escala = ', '.join(str(c) for c in h.choices), 'categórica'
            else:
                rango, escala = f'{h.low:g} a {h.high:g}', 'logarítmica' if h.log else 'lineal'
            filas.append({'Tarea': tarea, 'Modelo': modelo.name, 'Hiperparámetro': h.name, 'Rango': rango, 'Escala': escala})
pd.DataFrame(filas).set_index(['Tarea', 'Modelo'])
````

Para el desbalance de clases se compararon tres técnicas, siempre aplicadas solo sobre el tramo de entrenamiento de cada pliegue para no filtrar información. SMOTE genera ejemplos sintéticos de la clase minoritaria interpolando entre un caso y sus vecinos más cercanos (Chawla et al., 2002). ADASYN hace algo parecido pero crea más ejemplos donde la clase es más difícil de aprender (He et al., 2008). Los pesos de clase no crean datos: simplemente hacen que equivocarse en la clase minoritaria cueste más. Se aplicaron como pesos por observación, igual para todos los modelos; KNN no admite pesos, así que para ese modelo los pesos de clase equivalen a no balancear. Como en los primeros pliegues internos puede faltar una clase por completo (los primeros 13 meses de la serie son todos Alto), en esos casos no se remuestrea.

## Métodos de optimización de hiperparámetros

Los cuatro métodos evalúan cada combinación de hiperparámetros de la misma manera: reconstruyen a mano el pipeline completo (balanceo, escalado y modelo) dentro de cada pliegue interno y lo ajustan solo con el tramo de entrenamiento, de modo que el preprocesamiento nunca ve los datos de validación. El puntaje que se maximiza es el F1-macro en clasificación y el RMSE (con signo negativo) en regresión.

- **Grid Search** recorre una grilla de 4 valores por hiperparámetro numérico. Cuando la grilla completa tiene más de 40 combinaciones se elige al azar una submuestra de 40, que se recorre en el orden natural de la grilla (de los valores bajos a los altos de cada hiperparámetro); cuando tiene menos (Naive Bayes, Ridge y Lasso tienen 4, la regresión logística 8, KNN 16 y SVM 32) se recorre completa.
- **Random Search** evalúa 40 combinaciones muestreadas al azar del espacio de búsqueda, una alternativa que suele encontrar buenas configuraciones con menos evaluaciones que una grilla cuando pocos hiperparámetros importan de verdad (Bergstra & Bengio, 2012).
- **Optimización bayesiana**, con Optuna (Akiba et al., 2019), usa como modelo sustituto el estimador de Parzen de estructura de árbol (TPE; Bergstra et al., 2011). Después de 10 evaluaciones iniciales al azar, separa lo evaluado en un grupo bueno (el 10 % mejor) y otro malo, ajusta una densidad a cada uno, *l(x)* y *g(x)*, y propone el punto que maximiza *l(x)/g(x)*, que equivale a maximizar la mejora esperada. Esa función de adquisición favorece explorar cerca de donde ya se obtuvieron buenos resultados, sin abandonar del todo las zonas inciertas. Se escogió TPE porque maneja de forma natural un espacio mixto (enteros, continuos en escala logarítmica y categóricos) y no necesita invertir una matriz de covarianza, como haría un proceso gaussiano.
- **Algoritmo genético**, con DEAP (Fortin et al., 2012), representa cada hiperparámetro como un gen entre 0 y 1 que se traduce al rango real. La población es de 8 individuos y evoluciona con selección por torneo de tamaño 3, cruce por mezcla (`cxBlend`, con probabilidad 0.6), mutación gaussiana (probabilidad 0.3 por individuo y 0.3 por gen, desviación 0.15) y elitismo de un individuo, de modo que el mejor siempre pasa intacto a la siguiente generación. Solo se vuelven a evaluar los individuos que cambiaron, y la búsqueda termina al completar 40 evaluaciones. Las elecciones se justifican así: el torneo de tamaño 3 da una presión de selección moderada y no obliga a reescalar las aptitudes (que son F1 en clasificación y RMSE con signo negativo en regresión); el cruce por mezcla es el adecuado para genes continuos; la mutación gaussiana de desviación 0.15 (el 15 % del rango de cada gen) explora vecindades sin saltos bruscos; el elitismo evita perder la mejor solución, y una población de 8 permite repartir las 40 evaluaciones en unas 8 generaciones.

Para que la comparación de eficiencia sea justa, los cuatro disponen de unas 40 evaluaciones por búsqueda (menos en Grid cuando la grilla es más pequeña). Toda la corrida usa una única semilla global (42), que se propaga a NumPy, a scikit-learn, a XGBoost, a SMOTE/ADASYN, al muestreador de Optuna y al generador que usa DEAP; por eso repetir la corrida da exactamente los mismos resultados. Además de la tabla maestra, cada búsqueda guardó el puntaje y el tiempo de cada evaluación, la diversidad de la población en cada generación del algoritmo genético y las predicciones de cada pliegue externo.

### Calibración de las probabilidades

Un modelo puede ordenar bien los meses por riesgo y aun así dar probabilidades poco confiables: que diga "70 % de probabilidad de Alto" no significa que siete de cada diez meses con esa predicción sean realmente Alto. Para revisarlo se usan tres herramientas sobre la categoría Alto. El diagrama de confiabilidad agrupa las predicciones en 10 intervalos de probabilidad y compara la probabilidad media de cada intervalo con la frecuencia observada de meses Alto; si el modelo está bien calibrado, los puntos quedan sobre la diagonal. El Brier score (Brier, 1950) es el error cuadrático medio entre la probabilidad predicha y el resultado observado (0 o 1), y mezcla calibración con capacidad de discriminar. El error de calibración esperado, ECE (Guo et al., 2017), es el promedio, ponderado por el número de meses de cada intervalo, de la distancia entre la probabilidad predicha y la frecuencia observada. Para corregir los modelos mal calibrados se probaron dos recalibradores: el escalado de Platt (Platt, 1999), que ajusta una regresión logística sobre la probabilidad predicha, y la regresión isotónica (Zadrozny & Elkan, 2002), que ajusta una función escalonada creciente.

````python
curvas, filas = {}, []
for modelo, fila in mejores_clf.iterrows():
    pooled = oof_clf[modelo]
    calibracion = A.calibration_row(pooled)
    tabla = A.reliability_bins((pooled['y_true'] == A.ALTO).astype(int), pooled['y_proba'][:, A.ALTO], n_bins=10)
    curvas[etiquetas_modelo[modelo]] = (tabla, calibracion['brier_alto'], calibracion['ece_alto'])
    filas.append({'Modelo': etiquetas_modelo[modelo], 'Brier (Alto)': calibracion['brier_alto'],
                  'ECE (Alto)': calibracion['ece_alto'], 'Brier multiclase': calibracion['brier_multiclase']})
cuadro_calibracion = pd.DataFrame(filas).set_index('Modelo').sort_values('ECE (Alto)')
frecuencia_alto = float((oof_clf[mejores_clf.index[0]]['y_true'] == A.ALTO).mean())
print(f'Frecuencia de la categoría Alto: {frecuencia_alto:.3f}. Brier de una predicción constante igual a esa frecuencia: {frecuencia_alto * (1 - frecuencia_alto):.4f}')
cuadro_calibracion.round(4)
````

````python
fig = plots.reliability_grid(curvas)
plt.show()
````

La categoría Alto es rara (17 de 240 meses, el 7.1 %), así que el Brier score de esa categoría hay que compararlo con el de una predicción constante: darle siempre 7.1 % de probabilidad tendría un Brier de 0.066. Lo mejoran KNN (0.029), la regresión logística (0.047), Random Forest (0.049) y Naive Bayes (0.056); XGBoost (0.066) apenas lo iguala, y SVM (0.084) y el árbol de decisión (0.072) quedan peor que no predecir nada. En el ECE, KNN es el mejor calibrado (0.024) y XGBoost el peor (0.111), seguido de SVM (0.087), el árbol (0.082) y la regresión logística (0.077).

Los diagramas de confiabilidad hay que leerlos con cuidado. Casi todos los meses caen en el primer grupo (probabilidades cercanas a 0, la burbuja grande) y los grupos de probabilidad alta tienen unos pocos meses cada uno, así que sus puntos son muy ruidosos: un solo mes cambia la frecuencia observada de 0 a 1. Aun así se distinguen algunos patrones. Las probabilidades de XGBoost nunca pasan de 0.6, quedan comprimidas hacia la frecuencia base, algo esperable en un modelo que eligió una tasa de aprendizaje muy baja (0.007 en el último pliegue); las de SVM quedan casi siempre por debajo de la diagonal, es decir, sobreestiman el riesgo; el árbol de decisión da probabilidades de 0 o 1, por lo que su diagrama solo tiene tres puntos; y KNN y la regresión logística son los que más se acercan a la diagonal en los extremos.

````python
# Recalibración con escalado de Platt y con regresión isotónica: el modelo se entrena con el primer 75 % de cada
# tramo de entrenamiento y el recalibrador con el 25 % restante; la prueba solo se usa para evaluar.
filas = []
for modelo, fila in mejores_clf.iterrows():
    r = A.recalibration_experiment(modelo, fila['balancing'], fila['optimizer'], ds)
    filas.append({
        'Modelo': etiquetas_modelo[modelo],
        'Brier sin recalibrar': r['brier_sin_recalibrar'], 'Brier Platt': r['brier_platt'], 'Brier isotónica': r['brier_isotonica'],
        'ECE sin recalibrar': r['ece_sin_recalibrar'], 'ECE Platt': r['ece_platt'], 'ECE isotónica': r['ece_isotonica'],
    })
recalibracion = pd.DataFrame(filas).set_index('Modelo')
recalibracion.round(4)
````

El efecto de recalibrar depende de qué tan mal calibrado estuviera el modelo. Las mejoras más claras son las de los que tenían peor calibración: XGBoost baja su ECE de 0.126 a 0.057 con Platt y a 0.054 con la isotónica (y su Brier de 0.082 a 0.060 y 0.065), SVM de 0.133 a 0.070 con Platt (Brier de 0.111 a 0.074), Random Forest de 0.106 a 0.056 y la regresión logística de 0.078 a 0.049 con Platt y 0.045 con la isotónica. Naive Bayes mejora con Platt (ECE de 0.061 a 0.029) pero empeora con la isotónica (0.082).

En cambio, a KNN y al árbol de decisión recalibrar les hace daño: el ECE de KNN pasa de 0.047 a 0.080 con Platt y 0.074 con la isotónica, y el del árbol de 0.079 a 0.099 y 0.097. Eran los modelos que ya estaban razonablemente calibrados o cuyas probabilidades solo toman unos pocos valores, y el recalibrador se ajusta con apenas el 25 % final de cada tramo de entrenamiento, que tiene muy pocos meses Alto. Con tan pocos casos de la clase rara, Platt (dos parámetros) es más estable que la regresión isotónica (muchos escalones), que se sobreajusta, como se ve en Naive Bayes. La recomendación práctica es recalibrar solo los modelos mal calibrados (XGBoost, SVM, Random Forest y regresión logística), preferir Platt cuando hay pocos datos y comprobar siempre el efecto sobre una muestra que no se usó para ajustar el recalibrador.

## Resultados de regresión

En regresión se evalúan los siete modelos con el RMSE y el MAE, que son los errores del pronóstico en puntos porcentuales de desempleo. El RMSE castiga más los errores grandes (como los de 2020) y el MAE es el error absoluto promedio, así que mirar los dos ayuda a distinguir un modelo que falla poco en todas partes de uno que acierta casi siempre pero se equivoca mucho en unos pocos meses. Igual que en clasificación, para cada modelo se muestra su mejor optimizador según el RMSE promedio de los cinco pliegues externos, con la media ± desviación estándar entre pliegues.

Además de las métricas por pliegue, se juntaron los pronósticos de los cinco pliegues en una sola serie de 240 meses (de enero de 2006 a diciembre de 2025), todos hechos con información anterior. Con esa serie se calcularon las pruebas sobre los residuos, porque 240 observaciones dan mucha más potencia que los 48 meses de un solo pliegue. En el cuadro, la prueba de White (1980) contrasta que la varianza de los errores sea constante (se hace contra el valor predicho, porque los modelos no son lineales en las variables) y la de BDS (Brock et al., 1996) que los errores sean independientes entre sí.

````python
regresion = master[master['task'] == 'reg']
mejores_reg = A.best_per_model(master, 'reg')   # mejor combinación de cada modelo según el RMSE promedio

# Errores de pronóstico (observado menos predicho) de los 240 meses fuera de muestra, apilados
errores_reg = {m: A.pooled_regression_errors(f['run_id']) for m, f in mejores_reg.iterrows()}
supuestos = {m: A.residual_tests(e['error'], e['y_pred']) for m, e in errores_reg.items()}

def formato_p(valor):
    # los p-valores menores que 0.0001 se muestran como "< 0.0001" en vez de 0.0000
    return '< 0.0001' if valor < 0.0001 else f'{valor:.4f}'

cuadro_reg = pd.DataFrame({
    'Optimizador': mejores_reg['optimizer'].map(A.OPTIMIZER_LABELS),
    'RMSE': [A.mean_std(f, 'rmse') for _, f in mejores_reg.iterrows()],
    'MAE': [A.mean_std(f, 'mae') for _, f in mejores_reg.iterrows()],
    'p-valor (White)': [formato_p(supuestos[m]['white_p']) for m in mejores_reg.index],
    'p-valor (BDS)': [formato_p(supuestos[m]['bds_p']) for m in mejores_reg.index],
})
cuadro_reg.index = [A.MODEL_LABELS[m] for m in mejores_reg.index]
cuadro_reg
````

El mejor RMSE promedio entre pliegues es el de **Ridge (1.196 ± 0.552)**, seguido de cerca por SVR (1.213) y KNN (1.251); XGBoost (1.266) y Random Forest (1.274) vienen después, y el árbol de decisión (1.476) y Lasso (1.510) quedan de últimos. El MAE ordena casi igual, pero con menos distancia entre los cinco primeros (entre 0.87 y 0.89 puntos porcentuales). Que el RMSE sea bastante mayor que el MAE (1.2 contra 0.9) indica que hay errores grandes concentrados en pocos meses, que son los de 2020. La desviación entre pliegues es muy alta (de 0.5 a 0.9 puntos de RMSE), de modo que las diferencias entre los primeros cinco modelos están dentro del ruido.

Que el mejor modelo sea uno lineal tiene sentido en una serie tan persistente: el desempleo del mes siguiente se parece mucho al de este mes más una pequeña corrección, y los árboles, que predicen siempre dentro del rango de valores que vieron al entrenar, no extrapolan cuando la serie sale de ese rango. En las dos últimas columnas se ve que ningún modelo cumple los supuestos: el p-valor de la prueba de White es menor que 0.05 en los siete (la varianza de los errores no es constante) y el de BDS es menor que 0.0001 en los siete (los errores no son independientes). Esto se revisa en la sección de residuos.

## Comparación de los métodos de optimización

Comparar los cuatro métodos no es solo ver cuál terminó con la mejor métrica. Se miden tres cosas: cuánto cuesta cada método (evaluaciones, tiempo total y tiempo por evaluación), qué tan rápido se acerca cada uno a su mejor resultado (el desempeño en cualquier momento, es decir, el mejor puntaje encontrado hasta la evaluación *t* en función de *t*) y si las diferencias entre métodos son estadísticamente significativas. Para esto se guardó, en cada una de las 700 búsquedas (140 combinaciones por 5 pliegues), el puntaje y el tiempo de cada evaluación.

Como los problemas tienen escalas distintas (F1-macro en clasificación y RMSE en regresión), el desempeño en cualquier momento se mide con un arrepentimiento normalizado. En cada problema (un modelo, una técnica de balanceo y un pliegue externo) se toma como cero el mejor puntaje encontrado por cualquiera de los cuatro métodos y como uno el puntaje mediano de todas las evaluaciones; el arrepentimiento de un método tras *t* evaluaciones mide qué tan lejos queda del mejor puntaje conocido. Un método que llega pronto a cero es un método eficiente.

````python
# Historial de cada búsqueda: puntaje y tiempo de cada evaluación, y registro por generación del algoritmo genético
evaluaciones, generaciones, busquedas = OA.load_search_records(master)

costo = OA.cost_table(busquedas)
costo.index = [A.OPTIMIZER_LABELS[o] for o in costo.index]
costo.columns = ['Evaluaciones por búsqueda', 'Tiempo por búsqueda (s)', 'Tiempo total (min)', 'Costo del optimizador (%)', 'Tiempo por evaluación (s)']
costo.round(2)
````

Los cuatro métodos cuestan casi lo mismo por evaluación, así que comparar su calidad por presupuesto es justo. Random, el bayesiano y el genético hicieron 40 evaluaciones por búsqueda; Grid hizo en promedio 25.8, porque cuando la grilla completa es pequeña (Naive Bayes, Ridge y Lasso tienen 4 combinaciones y la regresión logística 8) se recorre entera y termina antes. El tiempo por evaluación va de 0.12 s (bayesiano) a 0.17 s (Grid), y el tiempo total de las búsquedas fue de 12.5 minutos para Grid, 14.6 para Random, 13.6 para el bayesiano y 15.9 para el genético. El costo del propio optimizador, es decir, lo que gasta en decidir qué evaluar sin contar el entrenamiento de los modelos, es despreciable en Random (0.02 %) y en el genético (0.07 %), y llega a 1.75 % en el bayesiano, que es el único que ajusta un modelo sustituto en cada iteración. Frente al costo de entrenar un Random Forest esto no cambia nada, pero sí importaría si cada evaluación tardara milisegundos.

La guía define el desempeño en cualquier momento como el mejor valor de la métrica encontrado hasta la evaluación *t*. ### Optimización multi-fidelidad

Todas las búsquedas anteriores evalúan cada configuración con el costo completo del modelo, incluso las que desde el comienzo se ven mal. El *Successive Halving* (Jamieson & Talwalkar, 2016) reparte mejor ese costo: empieza con muchas configuraciones entrenadas con un presupuesto pequeño, descarta las peores y le da más presupuesto a las que van quedando. Se implementó para Random Forest y XGBoost, que son los modelos que más tiempo consumen en la corrida de los 140 modelos.

La fidelidad es el número de árboles. Entrenar un bosque cuesta aproximadamente lo mismo por cada árbol, y con pocos árboles una configuración ya muestra si es prometedora, aunque su puntaje sea más ruidoso. El factor de reducción es 3: se arranca con 27 configuraciones entrenadas con 11 árboles, las 9 mejores pasan a 33 árboles, las 3 mejores a 100 y la mejor termina con 300. Con ese factor cada ronda cuesta casi lo mismo (unos 300 árboles por pliegue interno) y la búsqueda completa gasta 3582 árboles por pliegue externo, lo que equivale a entrenar cuatro modelos de 300 árboles. Para saber si ahorrar presupuesto así vale la pena se compara con dos variantes de Random Search: una que gasta el mismo costo en árboles (4 configuraciones completas) y la estándar de 40 evaluaciones, con entre 50 y 300 árboles. Cada método se repitió con tres semillas. Hyperband (Li et al., 2018) generaliza esta idea repitiendo el esquema con distintos números de configuraciones iniciales; aquí se usó la versión básica.

````python
mf = pd.read_csv(config.EXPERIMENTS_DIR / 'multifidelity.csv')
resumen_mf = mf.groupby(['tarea', 'modelo', 'metodo']).agg(
    metrica=('metrica_externa', 'mean'), desviacion=('metrica_externa', 'std'),
    tiempo=('tiempo_busqueda_s', 'mean'), costo=('costo_arboles_por_pliegue', 'mean'))
resumen_mf.columns = ['Métrica externa', 'Desviación entre semillas', 'Tiempo de búsqueda (s)', 'Costo en árboles por pliegue']
resumen_mf.index = pd.MultiIndex.from_tuples(
    [({'clf': 'Clasificación (F1-macro)', 'reg': 'Regresión (RMSE)'}[t], A.MODEL_LABELS[m], k) for t, m, k in resumen_mf.index],
    names=['Tarea', 'Modelo', 'Método'])
resumen_mf.round(3)
````

Successive Halving gasta 3582 árboles por pliegue externo, la sexta parte de lo que gasta Random Search con sus 40 evaluaciones (unos 21 mil árboles por pliegue, con 175 árboles de media por evaluación) y casi lo mismo que Random con el mismo costo (3600, cuatro configuraciones completas). Frente a esta última comparación, la justa, Successive Halving gana en los cuatro casos: sube el F1-macro de 0.484 a 0.492 en Random Forest y de 0.503 a 0.513 en XGBoost, y baja el RMSE de 1.374 a 1.270 en Random Forest y de 1.363 a 1.311 en XGBoost. Frente a Random Search estándar, que gasta seis veces más, consigue resultados parecidos: mejores en Random Forest (0.492 contra 0.483 de F1-macro y 1.270 contra 1.295 de RMSE) y peores por poco en XGBoost (0.513 contra 0.528 y 1.311 contra 1.306), con el tiempo de búsqueda dividido por entre 2.8 y 5.5 (18.5 s contra 102.3 s en Random Forest de clasificación, 12.9 s contra 35.7 s en XGBoost de clasificación, 14.6 s contra 78.9 s y 7.6 s contra 32.9 s en regresión). Conviene ser cautos: con tres semillas, la desviación entre semillas va de 0.005 a 0.083, así que diferencias de menos de 0.02 de F1-macro o de 0.05 de RMSE no son concluyentes. Lo que sí se puede afirmar es que reasignar el presupuesto hacia las configuraciones prometedoras permite acercarse a la calidad de la búsqueda completa con una fracción del costo.

Un detalle: el costo en árboles no es lo mismo que el tiempo. Con el mismo costo en árboles, Successive Halving tardó más que Random (18.5 s contra 16.6 s en Random Forest, 12.9 s contra 5.8 s en XGBoost de clasificación), porque entrena 40 modelos pequeños, de 11 y 33 árboles, en lugar de 4 grandes, y cada ajuste tiene un costo fijo de preparar los datos, balancear y escalar.

## Comparación estadística entre modelos

Con 140 modelos hay 9730 pares posibles. Si se aplicara una prueba como DeLong o Diebold-Mariano a cada uno, con un nivel de significancia de 0.05, solo por azar se esperarían cerca de 487 falsos positivos. Por eso no se comparan todos los pares: primero se hace una prueba global que dice si hay alguna diferencia entre los modelos, y solo después se pasa a comparaciones dirigidas entre un grupo pequeño de candidatos, corrigiendo los p-valores por comparaciones múltiples. Las pruebas de regresión se hicieron sobre los 240 errores mensuales fuera de muestra (no sobre las cinco métricas por pliegue), y las de clasificación sobre los pliegues externos y sobre las predicciones apiladas.

La secuencia en clasificación es la prueba ómnibus de Friedman (1937), el post-hoc de Nemenyi (1963) con el diagrama de diferencia crítica (Demšar, 2006), las comparaciones dirigidas de DeLong (DeLong et al., 1988) con el ajuste de Holm (1979) o el de Benjamini y Hochberg (1995), y el tamaño del efecto con el delta de Cliff (1993). En regresión es el Model Confidence Set (Hansen et al., 2011) sobre todos los modelos, el test de Giacomini y White (2006) entre los finalistas (el de Clark y West, 2007, no aplica porque ningún modelo es un caso particular de otro), el de Diebold y Mariano (1995) con la corrección de Harvey et al. (1997) como confirmación puntual, con intervalos de bootstrap estacionario (Politis & Romano, 1994), y la d de Cohen (1988) como tamaño del efecto. Para controlar el *data snooping* se agregan la prueba de habilidad predictiva superior de Hansen (2005) y el Reality Check de White (2000).

### Clasificación

````python
folds = A.load_folds()
f1_folds = folds[(folds['task'] == 'clf') & (folds['metric'] == 'f1_macro')]

# (a) Versión estricta: el mejor resultado de cada uno de los siete modelos en los cinco pliegues externos
estricto = pd.DataFrame({
    A.MODEL_LABELS[m]: folds[(folds['run_id'] == f['run_id']) & (folds['metric'] == 'f1_macro')].sort_values('fold')['value'].values
    for m, f in mejores_clf.iterrows()})
friedman_estricto = ST.friedman_test(estricto)

# (b) Versión ampliada: se usan las 16 variantes (balanceo y optimizador) de cada modelo en cada pliegue (80 bloques),
#     para no depender de haber elegido solo la mejor variante de cada modelo
ampliado = f1_folds.pivot_table(index=['balancing', 'optimizer', 'fold'], columns='model', values='value')
ampliado.columns = [A.MODEL_LABELS[m] for m in ampliado.columns]
friedman_ampliado = ST.friedman_test(ampliado)

# (c) El universo completo: los 112 modelos de clasificación en los cinco pliegues
universo = f1_folds.pivot_table(index='fold', columns='run_id', values='value')
friedman_universo = ST.friedman_test(universo)

pd.DataFrame({
    'Bloques': [len(estricto), len(ampliado), len(universo)],
    'Modelos comparados': [estricto.shape[1], ampliado.shape[1], universo.shape[1]],
    'Estadístico chi2': [friedman_estricto.statistic, friedman_ampliado.statistic, friedman_universo.statistic],
    'Grados de libertad': [friedman_estricto.df, friedman_ampliado.df, friedman_universo.df],
    'p-valor': [friedman_estricto.p_value, friedman_ampliado.p_value, friedman_universo.p_value],
}, index=['(a) Mejor variante de cada modelo', '(b) Todas las variantes de cada modelo', '(c) Los 112 modelos']).round(4)
````

La prueba de Friedman se aplicó de tres maneras, porque la respuesta depende de cuántos bloques se usen. Con los siete mejores modelos y los cinco pliegues (versión estricta) no se rechaza la igualdad: χ² = 3.68 con 6 grados de libertad y p = 0.72. Pero cinco bloques dan muy poca potencia, y además solo se está comparando a los ganadores de cada modelo. Cuando se usan las 16 variantes de cada modelo en cada pliegue (80 bloques) el resultado es claro: χ² = 47.36, 6 grados de libertad y p < 0.0001, es decir, no todos los modelos rinden igual. Con los 112 modelos y los cinco pliegues el resultado es marginal (χ² = 137.0, 111 grados de libertad, p = 0.048). Se concluye que sí hay diferencias entre los modelos cuando se mira el panorama completo, pero que con cinco pliegues no hay potencia para distinguir entre los mejores. Por eso el análisis post-hoc se hace con la versión ampliada.

````python
# Prueba post-hoc de Nemenyi y diagrama de diferencia crítica sobre la versión ampliada
cd_clf = ST.critical_difference(ampliado.shape[1], len(ampliado))
print('Rango promedio de cada modelo (menor es mejor):')
print(friedman_ampliado.avg_ranks.round(2).to_string())
print(f'Diferencia crítica de Nemenyi (alfa = 0.05): {cd_clf:.2f}')
fig, ax = plt.subplots(figsize=(10, 3.4))
ST.plot_cd_diagram(friedman_ampliado.avg_ranks, cd_clf, ax=ax, title='Diagrama de diferencia crítica (clasificación, 80 bloques)')
plt.show()

nemenyi_clf = ST.nemenyi_posthoc(ampliado)
print('p-valores de Nemenyi entre cada par de modelos:')
nemenyi_clf.round(3)
````

Con 7 modelos y 80 bloques, la diferencia crítica de Nemenyi es de 1.01 en rango promedio. Los rangos (menor es mejor) son 3.39 para Random Forest, 3.41 para KNN, 3.58 para XGBoost, 3.64 para Naive Bayes, 4.37 para el árbol de decisión, 4.57 para la regresión logística y 5.04 para SVM. Los cuatro primeros no se pueden distinguir entre sí, y el árbol de decisión tampoco se distingue de Random Forest (4.37 − 3.39 = 0.98, menos que 1.01). SVM es el único claramente peor que el primer grupo: se diferencia de Random Forest, KNN, XGBoost (p < 0.001) y Naive Bayes (p = 0.001), pero no del árbol ni de la regresión logística. La regresión logística es significativamente peor que KNN (p = 0.013) y que Random Forest (p = 0.010), aunque no que XGBoost (p = 0.059).

En otras palabras, KNN, Random Forest, XGBoost y Naive Bayes son equivalentes en rango, a pesar de que KNN tenga el mejor F1-macro en el cuadro de resultados: su ventaja en el mejor de sus 16 variantes no es consistente cuando se miran todas. Este diagrama reemplaza a las 21 comparaciones por pares entre los siete modelos, y para los 112 modelos reemplazaría a las 6216 que habría que hacer.

````python
# Comparaciones dirigidas con DeLong entre los tres modelos con mejor rango: AUC de la clase Alto sobre las predicciones agrupadas
mejores_rango = list(friedman_ampliado.avg_ranks.index[:3])
inverso = {A.MODEL_LABELS[m]: m for m in mejores_clf.index}
y_alto = oof_clf[inverso[mejores_rango[0]]]['y_true']
probabilidad_alto = {nombre: oof_clf[inverso[nombre]]['y_proba'][:, A.ALTO] for nombre in mejores_rango}

filas = []
for i in range(3):
    for j in range(i + 1, 3):
        a, b = mejores_rango[i], mejores_rango[j]
        r = ST.delong_test(y_alto, probabilidad_alto[a], probabilidad_alto[b], A.ALTO)
        delta = ST.cliffs_delta(estricto[a], estricto[b])
        filas.append({'Modelo A': a, 'Modelo B': b, 'AUC A': r['auc_a'], 'AUC B': r['auc_b'], 'Diferencia': r['diff'],
                      'IC 95 %': f"[{r['ci95'][0]:.3f}, {r['ci95'][1]:.3f}]", 'p crudo': r['p_value'],
                      'Delta de Cliff (F1 por pliegue)': delta})
delong = pd.DataFrame(filas)
delong['p ajustado (Holm)'] = ST.adjust_pvalues(delong['p crudo'].values, 'holm')
delong['p ajustado (BH)'] = ST.adjust_pvalues(delong['p crudo'].values, 'fdr_bh')
delong.set_index(['Modelo A', 'Modelo B']).round(3)
````

Entre los tres modelos con mejor rango (Random Forest, KNN y XGBoost) se compararon los AUC de la categoría Alto con la prueba de DeLong sobre las 240 predicciones apiladas. Las diferencias son pequeñas y ninguna es significativa: KNN supera a XGBoost en 0.029 de AUC (IC 95 % de −0.017 a 0.075, p = 0.212), Random Forest queda 0.035 por debajo de KNN (IC de −0.097 a 0.028, p = 0.275) y 0.006 por debajo de XGBoost (p = 0.844). Con la corrección de Holm los p-valores ajustados son 0.635, 0.844 y 0.635 (con Benjamini-Hochberg, 0.413, 0.844 y 0.413), así que no hay evidencia de que alguno de los tres distinga mejor los meses Alto. Hay que tener presente que DeLong supone observaciones independientes y los pronósticos mensuales están autocorrelacionados, de modo que los p-valores son, si acaso, demasiado optimistas, y aun así no resultan significativos.

El delta de Cliff sobre el F1-macro por pliegue sí muestra efectos medianos o grandes (−0.24 entre Random Forest y KNN, 0.44 entre Random Forest y XGBoost y 0.48 entre KNN y XGBoost), pero con solo cinco pliegues ese tamaño de efecto es muy inestable y apenas confirma el orden que ya mostraba el cuadro: KNN, Random Forest y XGBoost.

````python
# Intervalos de confianza al 95 % del AUC de la clase Alto y del RMSE de cada modelo de regresión, calculados de dos maneras:
# con el bootstrap estacionario (bloques de 12 meses en promedio), que respeta la dependencia entre meses seguidos, y con el
# BCa de remuestreo independiente, que la ignora y se incluye solo para ver cuánto cambia el intervalo.
from scipy.stats import bootstrap

def auc_alto(muestra):
    # AUC de la clase Alto; si una remuestra no tiene meses Alto el AUC no está definido
    return roc_auc_score(muestra[:, 0], muestra[:, 1]) if 0 < muestra[:, 0].sum() < len(muestra) else np.nan

def bca_auc(indicadora, probabilidad):
    r = bootstrap((indicadora, probabilidad), lambda a, b: auc_alto(np.column_stack([a, b])), paired=True, vectorized=False,
                  n_resamples=2000, method='BCa', random_state=config.SEED)
    return r.confidence_interval.low, r.confidence_interval.high

def bca_rmse(cuadrados):
    r = bootstrap((cuadrados,), lambda x, axis: np.sqrt(np.mean(x, axis=axis)), n_resamples=2000, method='BCa', random_state=config.SEED)
    return r.confidence_interval.low, r.confidence_interval.high

filas = []
for modelo, p in oof_clf.items():
    indicadora, probabilidad = (p['y_true'] == A.ALTO).astype(int), p['y_proba'][:, A.ALTO]
    punto, bajo, alto = ST.bootstrap_interval(np.column_stack([indicadora, probabilidad]), auc_alto, mean_block=12, n_boot=1000)
    bca_bajo, bca_alto = bca_auc(indicadora, probabilidad)
    filas.append({'Modelo': A.MODEL_LABELS[modelo], 'Métrica': 'AUC (Alto)', 'Estimación': punto, 'Estacionario inferior': bajo,
                  'Estacionario superior': alto, 'BCa inferior': bca_bajo, 'BCa superior': bca_alto})
referencias_rmse = {A.MODEL_LABELS[m]: errores_reg[m]['error'] ** 2 for m in mejores_reg.index}
for nombre in ['Persistencia', 'Estacional', 'SARIMA']:
    referencias_rmse[nombre] = ((referencias_reg['y_true'] - referencias_reg[nombre]) ** 2).values
for nombre, cuadrados in referencias_rmse.items():
    punto, bajo, alto = ST.bootstrap_interval(cuadrados, lambda a: np.sqrt(a.mean()), mean_block=12, n_boot=2000)
    bca_bajo, bca_alto = bca_rmse(cuadrados)
    filas.append({'Modelo': nombre, 'Métrica': 'RMSE', 'Estimación': punto, 'Estacionario inferior': bajo,
                  'Estacionario superior': alto, 'BCa inferior': bca_bajo, 'BCa superior': bca_alto})
intervalos = pd.DataFrame(filas).set_index(['Métrica', 'Modelo'])
intervalos.round(3)
````

Los intervalos del AUC de la categoría Alto son amplios porque solo hay 17 meses Alto: KNN tiene 0.951 con un intervalo de entre 0.77 y 0.99 (bootstrap estacionario) y los de Random Forest y XGBoost van de 0.82 a 0.97, así que los tres se solapan casi por completo, lo que concuerda con DeLong. El árbol de decisión (0.52, de 0.44 a 0.69) sí queda claramente por debajo de los tres, y SVM (0.67) también, pero con más incertidumbre. En RMSE los intervalos son todavía más anchos y se solapan todos, incluido el del SARIMA (0.995, de 0.50 a 1.57). Como el error total depende mucho de unos pocos meses de 2020, los intervalos son asimétricos y con un límite inferior muy bajo: las remuestras de bloques que no incluyen la pandemia dan errores mucho menores. Por eso comparar los intervalos de dos modelos por separado no es la forma de decidir cuál es mejor; las pruebas pareadas de la sección siguiente comparan los dos modelos mes a mes y son mucho más potentes (el intervalo de la diferencia SARIMA menos SVR, de −1.07 a −0.47, no incluye el cero aunque los intervalos individuales se solapen).

Los dos tipos de intervalo difieren de forma instructiva. El BCa de remuestreo independiente, que ignora que meses seguidos se parecen, es más estrecho (para el RMSE del SARIMA, de 0.70 a 1.56 frente a 0.50 a 1.57 del estacionario) y casi siempre tiene límites inferiores más altos. El estacionario preserva la dependencia mediante bloques de 12 meses de longitud media, es el apropiado para esta serie y por eso es el que se usa en el resto del capítulo; el BCa se muestra porque la guía lo recomienda cuando la métrica es asimétrica, y aquí coincide en lo cualitativo.

### Regresión

````python
# Errores de pronóstico de cada mes para los 28 modelos de regresión y las tres referencias; de ahí salen las pérdidas
errores_todos = {}
for _, f in regresion.iterrows():
    errores_todos[f"{A.MODEL_LABELS[f['model']]} ({A.OPTIMIZER_LABELS[f['optimizer']]})"] = A.pooled_regression_errors(f['run_id'])['error']
for nombre in ['Persistencia', 'Estacional', 'SARIMA']:
    errores_todos[nombre] = (referencias_reg['y_true'] - referencias_reg[nombre]).values
errores_todos = pd.DataFrame(errores_todos)
perdidas = errores_todos ** 2   # pérdida cuadrática de cada mes

# Model Confidence Set sobre el universo completo (bootstrap por bloques de 12 meses)
mcs = ST.model_confidence_set(perdidas, alpha=0.10, n_boot=5000, block=12)
mcs.columns = ['Modelo', 'Pérdida media (MSE)', 'Rango', 'p-valor MCS', 'En el conjunto de confianza']
print(f"Modelos en el conjunto de confianza (alfa = 0.10): {int(mcs['En el conjunto de confianza'].sum())} de {len(mcs)}")
mcs.head(15).set_index('Modelo').round(4)
````

````python
# Sensibilidad del MCS: con el error absoluto, menos dominado por unos pocos meses muy malos, con un universo reducido (la mejor
# variante de cada familia y las tres referencias) y con un nivel de significancia más holgado
familias = {}
for nombre in perdidas.columns:
    familias.setdefault(nombre.split(' (')[0], []).append(nombre)
reducido = [min(columnas, key=lambda c: perdidas[c].mean()) for columnas in familias.values()]
filas = []
for nombre, tabla in [('Error cuadrático', perdidas), ('Error absoluto', errores_todos.abs()),
                      ('Error cuadrático, universo reducido (10 modelos)', perdidas[reducido])]:
    for alfa in [0.10, 0.25]:
        r = ST.model_confidence_set(tabla, alpha=alfa, n_boot=5000, block=12)
        filas.append({'Pérdida': nombre, 'Nivel de significancia': alfa, 'Modelos en el conjunto': int(r['en_conjunto'].sum()), 'De': len(r)})
pd.DataFrame(filas).set_index(['Pérdida', 'Nivel de significancia'])
````

El resultado del MCS es poco informativo: no elimina a ningún modelo. Con la pérdida cuadrática y un nivel de 0.10, los 31 (los 28 de machine learning y las 3 referencias) quedan en el conjunto de confianza, incluso la regla del mismo mes del año anterior, cuyo MSE (4.44) es unas cuatro veces y media el del SARIMA (0.99). Con un nivel más holgado (0.25) sale uno, y con el error absoluto, menos dominado por los meses extremos, sale uno con 0.10 y salen ocho con 0.25. El universo reducido de 10 modelos (la mejor variante de cada familia y las tres referencias) tampoco elimina a ninguno, ni con 0.10 ni con 0.25. La razón es de potencia estadística: el MCS compara a cada modelo con el promedio de todo el grupo y tiene que superar el máximo de 31 estadísticos, y las pérdidas de estos modelos varían mucho de un mes a otro (unos pocos meses de 2020 aportan gran parte del error), de modo que el error estándar de la diferencia de pérdidas es del mismo orden que la diferencia misma.

La secuencia hay que leerla con cuidado, entonces: el MCS no permite reducir el conjunto de candidatos, y las comparaciones dirigidas se hacen entre finalistas elegidos por pérdida media. Lo que el MCS sí dice es que, con este tamaño de muestra, casi no hay evidencia para declarar a un modelo peor que el mejor. Las diferencias solo aparecen cuando se comparan pares de modelos con pruebas pareadas, que son más potentes porque cancelan la variación que comparten los dos modelos.

````python
# Como el MCS no elimina a ningún modelo, los finalistas se eligen por pérdida media: las dos mejores referencias y la mejor variante
# de las tres familias de machine learning con menor pérdida. Entre ellos no hay modelos anidados, así que se usa Giacomini-White
# (Clark-West no aplica) y Diebold-Mariano con la corrección de Harvey-Leybourne-Newbold, ambos sobre los errores mensuales fuera de muestra.
finalistas = ['SARIMA', 'Persistencia']
for nombre in perdidas.mean().sort_values().index:
    if nombre in ('SARIMA', 'Persistencia', 'Estacional') or nombre.split(' (')[0] in [f.split(' (')[0] for f in finalistas]:
        continue
    finalistas.append(nombre)
    if len(finalistas) == 5:
        break

filas = []
for i in range(len(finalistas)):
    for j in range(i + 1, len(finalistas)):
        a, b = finalistas[i], finalistas[j]
        gw = ST.giacomini_white_test(errores_todos[a].values, errores_todos[b].values)
        dm = ST.diebold_mariano_test(errores_todos[a].values, errores_todos[b].values)
        dif = (errores_todos[a] ** 2 - errores_todos[b] ** 2).values
        _, bajo, alto = ST.bootstrap_interval(dif, np.mean, mean_block=12, n_boot=2000)
        filas.append({'Modelo A': a, 'Modelo B': b, 'Diferencia media de pérdida': dif.mean(), 'IC 95 % (estacionario)': f'[{bajo:.3f}, {alto:.3f}]',
                      'p Diebold-Mariano (HLN)': dm['p_value'], 'p Giacomini-White': gw['p_value'], 'd de Cohen': ST.cohens_d_paired(dif)})
dirigidas = pd.DataFrame(filas)
dirigidas['p DM ajustado (Holm)'] = ST.adjust_pvalues(dirigidas['p Diebold-Mariano (HLN)'].values, 'holm')
dirigidas['p GW ajustado (Holm)'] = ST.adjust_pvalues(dirigidas['p Giacomini-White'].values, 'holm')
dirigidas.set_index(['Modelo A', 'Modelo B']).round(3)
````

Como el MCS no elimina a nadie, los finalistas se escogieron por pérdida media: las dos mejores referencias (SARIMA y persistencia) y la mejor variante de las tres familias de machine learning con menor pérdida (SVR, Ridge y KNN). Con la corrección de Holm sobre las 10 comparaciones, el patrón es nítido. El SARIMA es significativamente mejor que cada uno de los tres modelos de machine learning: la diferencia media de pérdida cuadrática es de −0.73 frente a SVR (IC 95 % estacionario de −1.07 a −0.47), de −0.75 frente a Ridge y de −0.89 frente a KNN, con p-valores de Diebold-Mariano (con la corrección HLN) y de Giacomini-White menores que 0.001 incluso después del ajuste, y tamaños de efecto pequeños (d de Cohen de −0.29 a −0.37). Entre la persistencia y los tres modelos de machine learning, y entre estos tres, no hay diferencias significativas: p de DM entre 0.29 y 0.84, p de Giacomini-White entre 0.07 y 0.24 sin ajustar, todos de 0.415 o más con el ajuste, y tamaños de efecto casi nulos (|d| entre 0.01 y 0.07). Es decir, ningún modelo de machine learning se distingue de repetir el último valor.

El caso SARIMA frente a persistencia es el interesante. El DM sin ajustar da p = 0.020 (diferencia de −0.68, IC de −1.17 a 0.13, que incluye el cero) y con Holm sube a 0.139, que ya no es significativo, mientras que Giacomini-White da p < 0.001. La explicación está en la tabla por periodos: antes y después de la pandemia el SARIMA gana con claridad, pero durante ella la persistencia lo supera (2.43 contra 3.03). Como la ventaja del SARIMA cambia de signo según el momento, el promedio incondicional que mide el DM la diluye, mientras que la prueba condicional de Giacomini-White sí detecta que la diferencia de pérdidas depende del estado de la serie. Es el caso que la guía menciona para preferir esta prueba.

````python
# Prueba SPA: ¿algún modelo de machine learning supera de verdad a la referencia, o es solo producto de haber probado 28?
candidatos = perdidas.drop(columns=['Persistencia', 'Estacional', 'SARIMA'])
filas = []
for referencia in ['SARIMA', 'Persistencia']:
    r = ST.spa_test(perdidas[referencia].values, candidatos, n_boot=5000, mean_block=12)
    filas.append({'Referencia': referencia, 'Mejor candidato': r['best_candidate'], 'Ganancia media de pérdida (MSE)': r['best_mean_loss_gain'],
                  'p SPA': r['p_consistent'], 'p Reality Check': r['p_upper']})
pd.DataFrame(filas).set_index('Referencia').round(4)
````

La prueba SPA responde la pregunta de fondo: ¿algún modelo de machine learning supera de verdad a las referencias o lo parecería solo por haber probado 28? Contra el SARIMA, el mejor candidato (SVR con Grid) tiene una pérdida mayor en 0.73, y los p-valores (0.60 del SPA y 0.90 del Reality Check) están lejos de la significancia: ningún modelo supera al SARIMA. Contra la persistencia, el mejor candidato (otra vez SVR con Grid) pierde por 0.05 y el p-valor es de 0.77 en las dos pruebas: tampoco ningún modelo la supera. Las dos pruebas coinciden cuando los candidatos están cerca de la referencia, como con la persistencia (0.77 en ambas), y se separan cuando casi todos son claramente peores, como con el SARIMA, donde el Reality Check, más conservador, da 0.90 y el SPA 0.60; ese es justo el sesgo que corrige Hansen.

## Interpretabilidad

Un modelo que predice bien no siempre es fácil de explicar, y en un problema como este, donde un pronóstico podría alimentar decisiones de política económica, interesa saber qué variables pesan en cada predicción. Para el mejor modelo de cada tarea se usaron los valores de Shapley (SHAP; Lundberg & Lee, 2017), que reparten la predicción entre las variables a partir de la teoría de juegos cooperativos (Shapley, 1953): las contribuciones sumadas al valor promedio del modelo reproducen la predicción (eficiencia), dos variables que aportan lo mismo reciben lo mismo (simetría) y las contribuciones de modelos combinados se suman (aditividad). Para árboles y modelos lineales el cálculo es exacto; para el resto, como KNN, se aproxima.

Primero se muestra la importancia global, con cada modelo entrenado con todos los meses; luego, dos meses en detalle por tarea (uno bien pronosticado y otro con un error grande), explicados con el modelo del pliegue que realmente los pronosticó; y al final se compara SHAP con LIME (Ribeiro et al., 2016) en XGBoost y se usa el mismo análisis para responder si los indicadores de las 13 áreas metropolitanas aportan información sobre el desempleo nacional.

### Importancia global con valores de Shapley

````python
import shap

# Modelo final de cada tarea entrenado con todos los meses, con los hiperparámetros del último pliegue externo
def modelo_final(tarea, modelo, fila):
    historial = A.load_history(fila['run_id'])
    parametros = historial['fold_results'][-1]['best_params']
    est, escalador, X_entrenamiento = A.fit_on_all(tarea, modelo, fila['balancing'], parametros, ds)
    return est, escalador, X_entrenamiento, parametros

mejor_clf = mejores_clf.index[0]
est_clf, esc_clf, X_clf, params_clf = modelo_final('clf', mejor_clf, mejores_clf.iloc[0])
explicador_clf, shap_clf = I.compute_shap(est_clf, esc_clf, X_clf, ds.X, 'clf')
importancia_clf = I.mean_abs_shap(shap_clf, ds.X.columns, A.ALTO)

est_reg, esc_reg, X_reg, params_reg = modelo_final('reg', mejor_reg, mejores_reg.iloc[0])
explicador_reg, shap_reg = I.compute_shap(est_reg, esc_reg, X_reg, ds.X, 'reg')
importancia_reg = I.mean_abs_shap(shap_reg, ds.X.columns)

print(f'Mejor clasificador: {A.MODEL_LABELS[mejor_clf]} con {params_clf}')
print(f'Mejor regresor: {A.MODEL_LABELS[mejor_reg]} con {params_reg}')
````

En los dos casos pesan sobre todo las variables que describen el nivel reciente del desempleo, aunque de forma distinta. En el KNN que estima la probabilidad de la categoría Alto la importancia está repartida entre muchas variables parecidas: el promedio móvil de 12 meses (10.4 % de la importancia total), el rezago de 12 meses (10.1 %), el promedio de 6 meses (9.1 %), el rezago de 6 meses (8.6 %) y la tasa de desempleo de las áreas (8.5 %) suman el 47 %. En todas ellas los valores altos (en rojo) empujan hacia arriba la probabilidad de Alto y los bajos hacia abajo, que es lo esperable: es más probable que un mes sea de crisis si el desempleo de los meses anteriores ya era alto. La estacionalidad también aparece, porque los valores altos de `mes_cos` (los meses de diciembre y enero, justo antes de los picos de comienzo de año) aumentan la probabilidad, y el índice de tendencia tiene el efecto contrario: a medida que avanza el tiempo la tasa baja.

En Ridge la variable más importante es la tasa de desempleo de las 13 áreas metropolitanas (13.7 %), seguida del rezago de 12 meses (12.3 %) y del último mes (12.2 %); el mes del año (`mes_cos` y `mes_sin`, 13.8 % entre las dos), la tasa de ocupación de las áreas (9.6 %) y la tendencia (9.6 %) completan el grupo que más pesa. Las tres variables de las áreas suman el 26.7 % de la importancia en Ridge (y el 13.9 % en el KNN), un primer indicio de que el desempleo de las ciudades principales sí aporta información sobre el nacional. Varias variables casi no cuentan: los rezagos de 2 y 3 meses y el promedio de 12 meses quedan en 0.0 %, 0.4 % y 0.1 %, porque Ridge, con una regularización fuerte (alpha de 70.9), reparte el peso entre variables muy correlacionadas y deja casi todo en las que mejor resumen el nivel y la estacionalidad.

### Dos observaciones en detalle

````python
# Para cada tarea se elige un mes que el modelo acertó bien y otro con un error grande, y se explica con el modelo del pliegue
# externo que lo pronosticó (así la explicación corresponde a una predicción realmente fuera de muestra).
def explicar_mes(tarea, modelo, fila, pos_en_oof):
    pooled = A.pooled_oof(fila['run_id'])
    mes = int(pooled['idx'][pos_en_oof])
    pliegue = int(pooled['fold'][pos_en_oof])
    ajuste = A.refit_fold(tarea, modelo, fila['balancing'], fila['optimizer'], pliegue, ds)
    explicador, valores = I.compute_shap(ajuste['est'], ajuste['scaler'], ajuste['X_train'], ajuste['X_test'], tarea)
    fila_en_prueba = int(np.where(ajuste['test_idx'] == mes)[0][0])
    entrada = ajuste['X_test_t'][fila_en_prueba:fila_en_prueba + 1]
    if tarea == 'clf':
        prediccion = A.M.expand_proba(ajuste['est'].predict_proba(entrada), ajuste['est'].classes_)[0, A.ALTO]
        clase = A.ALTO
    else:
        prediccion, clase = float(ajuste['est'].predict(entrada)[0]), None
    return I.local_explanation(explicador, valores, ajuste['X_test'], fila_en_prueba, clase), mes, prediccion

# Clasificación: un mes Alto bien clasificado y un mes Alto que el modelo no reconoció
p_clf = oof_clf[mejor_clf]
alto = np.where(p_clf['y_true'] == A.ALTO)[0]
acertado = alto[np.argmax(p_clf['y_proba'][alto, A.ALTO])]
fallado = alto[np.argmin(p_clf['y_proba'][alto, A.ALTO])]
# Regresión: el mes con el error más pequeño y el de error más grande
e_reg = errores_reg[mejor_reg]['error']
mejor_mes, peor_mes = int(np.argmin(np.abs(e_reg))), int(np.argmax(np.abs(e_reg)))

casos = [('clf', mejor_clf, mejores_clf.iloc[0], acertado, 'Clasificación: mes Alto bien clasificado'),
         ('clf', mejor_clf, mejores_clf.iloc[0], fallado, 'Clasificación: mes Alto no reconocido'),
         ('reg', mejor_reg, mejores_reg.iloc[0], mejor_mes, 'Regresión: error más pequeño'),
         ('reg', mejor_reg, mejores_reg.iloc[0], peor_mes, 'Regresión: error más grande')]
explicaciones = {}
for tarea, modelo, fila, posicion, titulo in casos:
    explicacion, mes, prediccion = explicar_mes(tarea, modelo, fila, posicion)
    explicaciones[titulo] = explicacion
    fecha = fechas_objetivo.iloc[mes].strftime('%Y-%m')
    if tarea == 'clf':
        detalle = f"observado {config.CLASS_LABELS[int(ds.y_clf.iloc[mes])]}, probabilidad de Alto {prediccion:.2f}"
    else:
        detalle = f"observado {ds.y_reg.iloc[mes]:.2f} %, predicho {prediccion:.2f} %"
    print(f'{titulo} ({fecha}): {detalle}')
````

En clasificación se explicó con el KNN un mes Alto bien clasificado (febrero de 2021, probabilidad de Alto de 1.00) y otro que no reconoció (enero de 2009, probabilidad de 0.00). En el primero todo empuja hacia arriba: la tasa de desempleo de las áreas (19.4 %), los promedios móviles de 3, 6 y 12 meses (entre 15.1 % y 17.0 %) y el último mes (17.6 %) están muy por encima de lo normal, y cada variable aporta entre 0.05 y 0.08 a la probabilidad. En el segundo ocurre lo contrario: el desempleo de diciembre de 2008 era de 10.8 % (y de 10.6 % el promedio de tres meses), nada en los últimos meses anunciaba una crisis, y todas las variables empujan la probabilidad hacia abajo (la de las áreas, la tendencia, el promedio de 12 meses y el rezago de 12 meses restan entre 0.05 y 0.07 cada una). El mes fue Alto por un salto estacional de enero que ningún modelo basado en la historia reciente podía prever: el modelo del primer pliegue solo había visto los años 2002-2005.

En regresión se explicó el Ridge para el mes con el error más pequeño (agosto de 2017: observado 9.52 %, predicho 9.51 %) y para el de error más grande (abril de 2020: observado 20.49 %, predicho 11.98 %). En el primero todas las variables bajan la predicción desde el promedio de 12.3 % hasta 9.5 %: el rezago de 12 meses (9.2 %) resta 0.73 puntos y el promedio móvil de 6 meses, el índice de tendencia, el último mes y el mes del año restan entre 0.25 y 0.28 cada uno. En el segundo, el Ridge sí reacciona a la caída de la tasa de ocupación de las áreas (52.6 %, que suma 0.48 puntos), pero esa señal se diluye entre el resto: el modelo ve un último mes de 13.2 % y un historial tranquilo, y como es lineal y con regularización fuerte no puede convertir una caída de la ocupación en un salto de 7 puntos, mayor que cualquiera de los que había en el entrenamiento (el máximo era de 4.7). La diferencia entre las dos explicaciones es que, cuando el error es pequeño, todas las variables apuntan en el mismo sentido y con intensidad parecida, y cuando es grande el patrón que explicaría el resultado no está en lo aprendido.

### SHAP frente a LIME en XGBoost

````python
# XGBoost entrenado con todos los meses: se compara la importancia media de SHAP con la que obtiene LIME sobre 40 meses repartidos
# a lo largo de la serie (LIME ajusta un modelo lineal local alrededor de cada mes).
muestra = ds.X.iloc[np.linspace(0, len(ds.X) - 1, 40).astype(int)]
filas, importancias = [], {}
for tarea, nombre in [('clf', 'Clasificación (clase Alto)'), ('reg', 'Regresión')]:
    registro = clasificacion if tarea == 'clf' else regresion
    fila = registro[registro['model'] == 'XGBoost'].sort_values('f1_macro_mean' if tarea == 'clf' else 'rmse_mean',
                                                                 ascending=(tarea == 'reg')).iloc[0]
    parametros = A.load_history(fila['run_id'])['fold_results'][-1]['best_params']
    est, esc, X_ent = A.fit_on_all(tarea, 'XGBoost', fila['balancing'], parametros, ds)
    clase = A.ALTO if tarea == 'clf' else None
    _, valores = I.compute_shap(est, esc, X_ent, muestra, tarea)
    shap_imp = I.mean_abs_shap(valores, ds.X.columns, clase)
    lime_imp = I.lime_importance(est, esc, X_ent, muestra, tarea, class_index=clase, n_instances=len(muestra))
    acuerdo = I.agreement(shap_imp, lime_imp, top=5)
    importancias[nombre] = (shap_imp, lime_imp)
    filas.append({'Tarea': nombre, 'Correlación de Spearman': acuerdo['spearman'],
                  'Variables en común entre las 5 primeras': acuerdo['top_en_comun']})
for nombre, (shap_imp, lime_imp) in importancias.items():
    print(f'{nombre}\n  Primeras 5 según SHAP: {", ".join(shap_imp.index[:5])}\n  Primeras 5 según LIME: {", ".join(lime_imp.index[:5])}')
pd.DataFrame(filas).set_index('Tarea').round(3)
````

````python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
for ax, (nombre, (shap_imp, lime_imp)) in zip(axes, importancias.items()):
    comparacion = pd.concat([(shap_imp / shap_imp.sum()).rename('SHAP'), (lime_imp / lime_imp.sum()).rename('LIME')], axis=1)
    comparacion = comparacion.loc[comparacion['SHAP'].sort_values(ascending=False).index[:8]][::-1]
    posiciones = np.arange(len(comparacion))
    ax.barh(posiciones - 0.2, comparacion['SHAP'], height=0.4, color=plots.BLUE, label='SHAP')
    ax.barh(posiciones + 0.2, comparacion['LIME'], height=0.4, color=plots.RED, label='LIME')
    ax.set_yticks(posiciones, comparacion.index, fontsize=8)
    ax.set_xlabel('Importancia relativa')
    ax.set_title(f'XGBoost: {nombre}')
    ax.legend()
plt.tight_layout()
plt.show()
````

Para XGBoost, SHAP y LIME dan rankings muy parecidos: la correlación de Spearman entre las dos importancias es de 0.95 en clasificación (categoría Alto) y de 0.82 en regresión, y las cinco variables principales son las mismas con los dos métodos en las dos tareas. En clasificación encabezan la lista la tasa de desempleo de las áreas y el promedio móvil de 3 meses (con el orden intercambiado entre métodos), seguidos de los promedios y rezagos de 6 y 12 meses; en regresión, el último mes, la tasa de desempleo de las áreas, el promedio de 3 meses y el mes del año, con la tasa de participación de las áreas en quinto lugar. Las diferencias aparecen en los detalles: en regresión, LIME le da más peso a la tasa de desempleo de las áreas (0.19 frente a 0.14 de SHAP) y menos al mes del año (0.11 contra 0.14), y en clasificación le da menos a la tasa de desempleo de las áreas (0.20 contra 0.25) y más al promedio de 12 meses (0.06 contra 0.04).

Son las diferencias que cabe esperar de dos métodos distintos. SHAP reparte la predicción entre las variables con valores de Shapley, que para árboles se calculan de forma exacta y cumplen eficiencia, simetría y aditividad; LIME ajusta una regresión lineal ponderada alrededor de cada mes con observaciones perturbadas al azar, así que su explicación es local, depende del tamaño de la vecindad y del muestreo, y solo captura lo lineal. Los dos pueden divergir cuando las variables están muy correlacionadas (aquí los rezagos y los promedios móviles de la tasa lo están, y el reparto del peso entre ellas cambia de un método a otro), cuando el modelo tiene interacciones o saltos fuertes cerca de un umbral que una recta no representa, y cuando se explica una observación aislada en lugar del promedio de muchas, como se hizo aquí con 40 meses. Que coincidan en el ranking promedio es una buena señal de que la explicación no depende del método, pero no garantiza que coincidan en cada mes.

## Optimización computacional

Con 288 observaciones y 17 variables el costo computacional no es un problema: la corrida completa de los 140 modelos tomó 57 minutos sin paralelizar nada. Aun así, conviene entender cómo escalaría cada modelo si los datos fueran mucho más grandes, y qué se gana al usar la variante pensada para ese caso. Esta sección hace tres cosas: contrasta la complejidad teórica de cada modelo con el tiempo medido en el proyecto, mide el escalamiento con n y con p sobre datos sintéticos de tamaño controlado (cada medición se repitió tres veces, o dos en los modelos más lentos, y se reporta la mediana), y compara la variante estándar de cada modelo con la optimizada: KNN exhaustivo frente a KD-Tree y FAISS (Johnson et al., 2019), Ridge con solver directo frente a SAGA (Defazio et al., 2014), Naive Bayes con `fit` frente a `partial_fit`, XGBoost con método exacto frente a histogramas y SVM con núcleo frente a `LinearSVC` y gradiente estocástico. Todas las mediciones de esta sección se hicieron con BLAS y OpenMP limitados a un hilo: en una primera versión el Lasso saltó de 0.01 s a 1 s entre 10 mil y 20 mil observaciones, y resultó ser un artefacto de los 12 hilos de OpenBLAS compitiendo con otros procesos y no un comportamiento del algoritmo (con un hilo el tiempo crece de forma lineal).

### Complejidad teórica y tiempo medido en el proyecto

| Modelo | Entrenamiento | Inferencia | Variante pensada para datos grandes |
|---|---|---|---|
| KNN | $O(1)$ (solo guarda los datos) | $O(n \cdot p)$ por consulta con búsqueda exhaustiva | KD-Tree o Ball-Tree, $O(p \log n)$ en pocas dimensiones; FAISS para vecinos aproximados |
| Ridge | $O(n p^2 + p^3)$ con la ecuación normal | $O(p)$ | SAGA, con costo lineal en $n$ y $p$ por iteración |
| Lasso | $O(n p \cdot \text{iter})$ por descenso de coordenadas | $O(p)$ | gradiente estocástico |
| Naive Bayes | $O(n p)$ | $O(p)$ | `partial_fit`, que aprende por lotes sin cargar todo en memoria |
| Regresión logística | $O(n p \cdot \text{iter})$ | $O(p)$ | SAGA |
| Árbol de decisión | $O(n p \log n)$ | $O(\text{profundidad})$ | |
| Random Forest | $O(B \cdot n p \log n)$, con $B$ árboles | $O(B \cdot \text{profundidad})$ | árboles en paralelo con `n_jobs` |
| XGBoost | $O(B \cdot n p \log n)$ con el método exacto | $O(B \cdot \text{profundidad})$ | histogramas, $O(B \cdot n p)$, y parada temprana |
| SVM y SVR con núcleo | entre $O(n^2 p)$ y $O(n^3)$ | $O(n_{vs} \cdot p)$, con $n_{vs}$ vectores de soporte | `LinearSVC` o gradiente estocástico, $O(n p)$ |

````python
# Tiempo medido en la corrida real (n = 239 meses de entrenamiento por pliegue en promedio), por familia de modelos
tiempos = master.groupby(['task', 'model']).agg(
    ajuste_final=('final_fit_time_s', 'mean'), inferencia=('predict_time_s', 'mean'),
    busqueda=('search_time_s', 'sum'), evaluaciones=('total_evaluations', 'sum'))
tiempos['ms por evaluación'] = 1000 * tiempos['busqueda'] / tiempos['evaluaciones']
tiempos['ajuste_final'] *= 1000
tiempos['inferencia'] *= 1000
tiempos['busqueda'] /= 60
tiempos.index = pd.MultiIndex.from_tuples(
    [({'clf': 'Clasificación', 'reg': 'Regresión'}[t], A.MODEL_LABELS[m]) for t, m in tiempos.index], names=['Tarea', 'Modelo'])
tiempos = tiempos.rename(columns={'ajuste_final': 'Ajuste final (ms)', 'inferencia': 'Inferencia de 48 meses (ms)',
                                  'busqueda': 'Búsqueda total (min)', 'evaluaciones': 'Evaluaciones'})
tiempos.sort_values('Búsqueda total (min)', ascending=False).round(2)
````

La búsqueda de los 140 modelos sumó 56.7 minutos, y el 82 % de ese tiempo se fue en Random Forest y XGBoost. Solo el Random Forest de clasificación tomó 25.8 minutos (484 ms por evaluación, y cada evaluación entrena tres modelos, uno por pliegue interno), seguido de XGBoost de clasificación con 13.0 minutos y de Random Forest de regresión con 5.6. Los demás cuestan 4.1 minutos o menos cada uno (la regresión logística es la más cara de ellos), y Ridge, Lasso, KNN y el árbol de regresión menos de 15 segundos. Esto es lo que anticipaba la tabla de complejidad: los métodos de conjunto multiplican el costo de un árbol por el número de árboles (hasta 300 en el espacio de búsqueda), mientras que los modelos lineales pasan pocas veces por los datos. También se ve la diferencia entre el aprendizaje perezoso y el ansioso: KNN casi no cuesta entrenarlo (2.8 ms) y predecir 48 meses le toma más (7.1 ms), mientras que en el Random Forest ocurre al revés (175 ms de ajuste y 17 ms de predicción). Con solo 239 meses de entrenamiento los tiempos absolutos son pequeños, y por eso el efecto de las variantes optimizadas solo se puede estudiar con datos sintéticos más grandes, como se hace a continuación.

### Escalamiento con el tamaño de los datos

````python
escalamiento = pd.read_csv(config.EXPERIMENTS_DIR / 'complexity_scaling.csv')

fig, ejes = plt.subplots(2, 4, figsize=(18, 8))
plots.plot_scaling(escalamiento, 'KNN', x='n', y='inferencia_s', ax=ejes[0, 0], title='KNN: inferencia frente a n')
plots.plot_scaling(escalamiento, 'KNN (en p)', x='p', y='inferencia_s', ax=ejes[0, 1], title='KNN: inferencia frente a p')
plots.plot_scaling(escalamiento, 'Ridge/Lasso', x='n', y='entrenamiento_s', ax=ejes[0, 2], title='Ridge y Lasso: entrenamiento frente a n')
plots.plot_scaling(escalamiento, 'Ridge/Lasso (en p)', x='p', y='entrenamiento_s', ax=ejes[0, 3], title='Ridge y Lasso: entrenamiento frente a p')
plots.plot_scaling(escalamiento, 'Naive Bayes', x='n', y='entrenamiento_s', ax=ejes[1, 0], title='Naive Bayes: entrenamiento frente a n')
plots.plot_scaling(escalamiento, 'XGBoost', x='n', y='entrenamiento_s', ax=ejes[1, 1], title='XGBoost: entrenamiento frente a n')
plots.plot_scaling(escalamiento, 'SVM', x='n', y='entrenamiento_s', ax=ejes[1, 2], title='SVM: entrenamiento frente a n')
plots.plot_scaling(escalamiento, 'SVM', x='n', y='inferencia_s', ax=ejes[1, 3], title='SVM: inferencia frente a n')
plt.tight_layout()
plt.show()
````

````python
# Exponente empírico: pendiente de log(tiempo) frente a log(n) (o log(p)); un exponente cercano a 1 indica crecimiento lineal
filas = []
for familia, x in [('KNN', 'n'), ('KNN (en p)', 'p'), ('Ridge/Lasso', 'n'), ('Ridge/Lasso (en p)', 'p'),
                   ('Naive Bayes', 'n'), ('XGBoost', 'n'), ('SVM', 'n')]:
    sub = escalamiento[escalamiento['familia'] == familia]
    for variante, grupo in sub.groupby('variante', sort=False):
        fila = {'Familia': familia, 'Variante': variante, 'Variable': x}
        for columna, nombre in [('entrenamiento_s', 'Exponente de entrenamiento'), ('inferencia_s', 'Exponente de inferencia')]:
            valido = grupo[(grupo[columna] > 0) & (grupo[x] > 0)]
            fila[nombre] = np.polyfit(np.log(valido[x]), np.log(valido[columna]), 1)[0] if len(valido) >= 3 else np.nan
        filas.append(fila)
exponentes = pd.DataFrame(filas).set_index(['Familia', 'Variante'])
exponentes.round(2)
````

La mayoría de los exponentes medidos coincide con lo que dice la teoría. En KNN, la búsqueda exhaustiva predice con un costo casi lineal en n (exponente de 0.97) y el KD-Tree crece mucho más despacio (0.25, cerca del logarítmico), a cambio de un índice que sí cuesta construir casi n log n (1.15 en entrenamiento). Aun así, con 8 variables el KD-Tree solo empieza a ganarle a la búsqueda exhaustiva a partir de unas 16 mil a 32 mil observaciones (con 32 mil predice en 0.15 s y la exhaustiva en 0.21 s; con 2 mil tarda 0.076 s contra 0.014 s). Y cuando crece la dimensión pierde su ventaja por completo: con 80 variables el KD-Tree tarda 1.9 s frente a 0.1 s de la búsqueda exhaustiva, que es la maldición de la dimensionalidad. FAISS exacto escala igual que la búsqueda exhaustiva (0.99), solo que más rápido, y la versión aproximada (IVF) crece algo menos (0.81).

Ridge y Lasso crecen de forma lineal en n y en p, como dice el costo por iteración de O(n·p) (exponentes entre 0.9 y 1.1). El solver directo (Cholesky) tiene un exponente de 0.91 en n y de 1.04 en p; la teoría dice que debería crecer con p² por el producto XᵀX, pero entre 10 y 400 variables su costo está dominado por recorrer la matriz y no por el término cuadrático, que solo se notaría con muchas más variables. XGBoost con el método exacto crece como n log n (1.15) y con histogramas bastante menos (0.60); su inferencia no crece con n (exponente cercano a cero) porque depende de los árboles y no de los datos de entrenamiento. En SVM con núcleo RBF el entrenamiento crece de forma claramente superlineal (1.67, por debajo del cuadrático teórico porque el costo depende de cuántos vectores de soporte quedan) y la inferencia también crece con n (0.73), porque cada predicción se compara con los vectores de soporte; LinearSVC (1.02) y el gradiente estocástico (0.90) son lineales y su inferencia no depende de n. En Naive Bayes, `fit` y `partial_fit` son igual de lineales (1.00 y 0.97): aprender por lotes no cambia el orden del tiempo, cambia la memoria, como se ve en la tabla siguiente.

La tabla muestra que "optimizado" no siempre significa "más rápido": depende del tamaño de los datos y de qué se sacrifica. En KNN con 16 mil observaciones y 8 variables, el KD-Tree no ayuda (predice en 0.135 s contra 0.142 s de la búsqueda exhaustiva y construir el índice cuesta 40 veces más), pero FAISS aproximado predice 6.1 veces más rápido (0.023 s) con casi la misma exactitud (0.9575 contra 0.9585): el compromiso entre exactitud y velocidad de la aproximación es aquí de una milésima de accuracy. En Ridge, SAGA es 8.4 veces más lento que Cholesky (0.186 s contra 0.022 s) con el mismo RMSE (10.2439 y 10.2440); solo compensaría con muchas más variables o con matrices dispersas, que no se probaron. En Naive Bayes, `partial_fit` tarda lo mismo pero usa 8.9 MB en lugar de 508 MB, 57 veces menos memoria, que es justo su razón de ser: aprender cuando los datos no caben en memoria. XGBoost con histogramas entrena 15.6 veces más rápido (2.5 s contra 38.7 s con 40 mil observaciones) y pierde solo 0.0005 de accuracy (de 0.9815 a 0.9810), la mejor relación de toda la tabla, a cambio de algo más de memoria (16.4 MB contra 12.4 MB) por las estructuras de los histogramas. Con parada temprana, XGBoost se detuvo en 78 de los 300 árboles y entrenó 2.4 veces más rápido (1.55 s contra 3.78 s) con el mismo accuracy (0.9430 contra 0.9435).

En SVM, LinearSVC y el gradiente estocástico entrenan 29 y 33 veces más rápido, predicen en 0.3 milisegundos en vez de 0.43 s y usan 2.7 MB y 0.02 MB en lugar de 34.9 MB, pero el accuracy cae de 0.9245 a 0.839 y a 0.816. Es el único caso donde la variante optimizada paga un precio claro en desempeño, porque el problema sintético no es linealmente separable del todo. Y Random Forest con todos los núcleos (`n_jobs=-1`) entrena 7.6 veces más rápido con exactamente el mismo resultado (0.9315). En resumen, las mejoras que no cambian el modelo (histogramas, paralelismo, parada temprana) son prácticamente gratis, las que lo aproximan (FAISS, SVM lineal) cambian velocidad por exactitud en una proporción que hay que medir caso por caso, y las que solo ahorran memoria (`partial_fit`, SGD) son útiles cuando los datos no caben, no cuando sobra memoria.

### Paralelización

El trabajo de este proyecto se puede repartir de dos maneras: entre combinaciones (cada una de las 140 es independiente de las demás) y entre pliegues externos (los cinco pliegues de una misma combinación también lo son). Con `joblib` se midieron las dos. Para la primera se corrieron, en serie y en paralelo, 24 combinaciones baratas (KNN, Ridge y Lasso con sus cuatro optimizadores) con el backend `loky`, que reparte el trabajo en procesos, y con `threading`, que lo reparte en hilos que comparten memoria pero quedan limitados por el GIL de Python. Para la segunda se corrió una combinación pesada, Random Forest con búsqueda aleatoria, repartiendo sus cinco pliegues externos en cinco procesos. En todos los casos se comprobó que el resultado paralelo fuera idéntico al serial, porque paralelizar no puede cambiar las métricas. En estas mediciones BLAS y OpenMP se limitaron a un hilo para que lo único que se paralelice sea lo que se pide con `n_jobs`.

````python
paralelo = pd.read_csv(config.EXPERIMENTS_DIR / 'parallel_speedup.csv')
paralelo = paralelo.rename(columns={'parte': 'Qué se reparte', 'backend': 'Backend de joblib', 'trabajadores': 'Trabajadores',
                                    'tiempo_s': 'Tiempo (s)', 'aceleracion': 'Aceleración', 'eficiencia': 'Eficiencia',
                                    'resultado_identico': 'Resultado idéntico al serial'})
paralelo.set_index(['Qué se reparte', 'Backend de joblib', 'Trabajadores']).round(2)
````

Repartir las combinaciones entre procesos acelera de forma clara pero no proporcional: con 2 procesos se gana 1.73 veces (una eficiencia de 86 %), con 4 procesos 3.10 veces (78 %), con 8 procesos 4.12 veces (52 %) y con 12 solo 4.34 veces (36 %). Pasar de 8 a 12 casi no ayuda porque el procesador tiene 6 núcleos físicos (los otros 6 hilos comparten núcleo con ellos), porque cada proceso nuevo tiene que importar las librerías y recibir los datos, y porque las 24 combinaciones no cuestan lo mismo, así que al final los trabajadores esperan a la más lenta. Repartir los cinco pliegues externos de una combinación pesada (Random Forest con búsqueda aleatoria) entre 5 procesos baja el tiempo de 93.1 s a 25.7 s (3.63 veces, 73 % de eficiencia), también con resultados idénticos a los seriales. Si la aceleración de 4.3 veces se mantuviera en la corrida completa, los 57 minutos se reducirían a unos 13.

El backend `threading`, en cambio, empeora las cosas: con 4 y 8 hilos el tiempo (83.6 s y 88.1 s) es mayor que el serial (73.4 s), es decir, aceleraciones de 0.88 y 0.83. El código de los optimizadores es Python puro y el GIL deja avanzar a un solo hilo a la vez, así que coordinar los hilos cuesta más de lo que se gana. Además los hilos cambian el resultado, como lo indica la última columna. Una prueba aparte con 16 de las combinaciones (`scripts/diagnostico_hilos.py`) mostró que solo cambian las del algoritmo genético: DEAP usa el generador aleatorio global de Python, que los hilos comparten y que cada búsqueda genética reinicia con la semilla al empezar, de modo que las secuencias de números se mezclan entre hilos. Por eso cuáles combinaciones cambian, y por cuánto, varía de una ejecución a otra (entre 0.0001 y 0.011 en las dos veces que se corrió el diagnóstico), lo típico de una condición de carrera. Con procesos (`loky`) cada trabajador tiene su propio generador y las 16 combinaciones dieron exactamente lo mismo, por lo que `loky` es el backend adecuado para este proyecto.

No se usó Dask ni Ray porque no se justifican aquí: las 140 combinaciones suman 57 minutos de cómputo, los datos son de 288 filas y `joblib` ya da una aceleración de 3.6 a 4.3 veces en un solo computador. Esas herramientas valen la pena cuando el trabajo no cabe en una máquina, con muchas más combinaciones pesadas o datos de millones de filas. Tampoco se pudo medir la aceleración por GPU de XGBoost (`device='cuda'`): el computador tiene tarjetas AMD Radeon, que no tienen CUDA, así que no hay un factor de aceleración que reportar. Con 239 observaciones por entrenamiento sería de esperar que la GPU no ayudara, porque copiar los datos a la tarjeta costaría más que el entrenamiento mismo (el ajuste final de XGBoost toma unos 118 ms), pero esa es una expectativa y no una medición.

## Conclusiones del modelado con machine learning

Este capítulo comparó siete modelos de clasificación y siete de regresión con cuatro técnicas de balanceo (solo en clasificación) y cuatro métodos de optimización de hiperparámetros, es decir, 140 modelos evaluados con una validación anidada que respeta el orden del tiempo. El resultado principal es poco halagador para el machine learning y por eso mismo es el más útil: **ningún modelo de regresión superó al SARIMA(0,1,1)(0,1,1)[12] del capítulo anterior** (RMSE de 0.995 frente a 1.314 del mejor, SVR) ni, con significancia, a la simple persistencia (1.293), y en clasificación **solo KNN superó a la regla de repetir la categoría del mes actual** (F1-macro de 0.815 frente a 0.752). La razón se ve en los residuos y en los pronósticos: la serie es en buena parte estacionalidad y persistencia, que el SARIMA modela de forma explícita, mientras que los modelos de machine learning, con 288 observaciones y 17 variables muy correlacionadas, aprenden esa estructura solo a medias (los residuos conservan autocorrelación y un pico en el rezago 12) y fallan donde la serie se rompe, como en abril de 2020.

Dentro de lo que sí se pudo comparar, en clasificación KNN, Random Forest, XGBoost y Naive Bayes son equivalentes en rango según Nemenyi y SVM es el peor; ADASYN es la técnica de balanceo más segura (mejora media de 0.049 de F1-macro, delta de Cliff de 0.40), aunque no ayuda a todos los modelos y a algunos, como el árbol de decisión y XGBoost, no les sirve ninguna; y las probabilidades de XGBoost, SVM, Random Forest y regresión logística salen mal calibradas y mejoran con el escalado de Platt (el ECE de XGBoost baja de 0.126 a 0.057), mientras que recalibrar KNN o el árbol las empeora. En regresión, entre los modelos de machine learning no hay diferencias significativas; los árboles son los mejores antes de la pandemia y los peores durante ella, y los regularizados (Ridge y SVR) son los más robustos al choque.

Entre los métodos de optimización, el bayesiano fue el más eficiente: alcanza el 10 % del mejor puntaje con una mediana de 15 evaluaciones (Random necesita 35.5, el genético 39 y Grid casi nunca llega), tiene el área bajo la curva de arrepentimiento más baja (0.28 contra 0.36, 0.38 y 0.49) y es significativamente mejor que Grid, aunque no que Random Search ni que el genético. El genético sufre convergencia prematura, porque su población pierde casi 90 % de su diversidad en siete generaciones. La calidad final del modelo depende poco del optimizador, y Successive Halving consigue una calidad parecida a la de Random Search con la sexta parte del costo en árboles. Los resultados son reproducibles con la semilla global, y la aleatoriedad de los modelos mueve mucho menos el resultado (desviaciones de hasta 0.06 de F1-macro y 0.04 de RMSE) que la que hay entre pliegues.

En el costo computacional, Random Forest y XGBoost consumen el 82 % de los 57 minutos de la corrida. Las mejoras que no cambian el modelo son casi gratis (los histogramas de XGBoost lo aceleran 15.6 veces, la parada temprana 2.4 y repartir pliegues entre procesos entre 3.6 y 4.3), mientras que las que lo aproximan cambian exactitud por velocidad (LinearSVC es 29 veces más rápido pero pierde 8.6 puntos porcentuales de accuracy, y FAISS aproximado es 6.1 veces más rápido y pierde una milésima). Los hilos de Python no sirven para este trabajo, porque el GIL los frena y, además, cambian el resultado del algoritmo genético, y no fue posible medir la aceleración con GPU porque el equipo no tiene tarjeta con CUDA.

Respecto a las hipótesis del proyecto, la primera hipótesis secundaria se sostiene: la tasa de desempleo de las 13 áreas metropolitanas es la variable más importante en Ridge (13.7 % de la importancia) y en XGBoost para la categoría Alto, y quitar los tres indicadores de área empeora el RMSE en cinco de los siete modelos (hasta 9 % en XGBoost), aunque la evidencia es moderada. La ruptura de la pandemia (segunda hipótesis) se ve en el desempeño: el error se multiplica entre 2 y 6 veces, según el modelo, entre los periodos tranquilos y 2020-2021 (de 0.51 a 3.03 en el SARIMA, de 0.83 a 4.93 en XGBoost), y ningún modelo anticipa el salto de abril de 2020. La estacionalidad (tercera hipótesis) es modelable, pero el SARIMA la modela mucho mejor que los modelos de machine learning con la codificación de seno y coseno que se usó.

Hay límites que condicionan todo lo anterior. Con 288 meses y solo 17 meses Alto en las predicciones fuera de muestra, cualquier conclusión sobre esa categoría tiene mucha incertidumbre; los cortes de las categorías (10 % y 14 %) son una decisión propia, y las cifras de la mejor variante de cada modelo son algo optimistas porque se escogió la mejor de 16. Los pasos siguientes que más prometen son enriquecer las variables de estacionalidad (más armónicos o variables por mes) y trabajar con diferencias en lugar de niveles, combinar el SARIMA con un modelo de machine learning sobre sus residuos, incorporar variables externas (por ejemplo, de actividad económica o de precios), y probar poblaciones más grandes en el algoritmo genético.
